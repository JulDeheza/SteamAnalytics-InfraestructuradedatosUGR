
## 4. Construcción y auditoría del corpus principal

### 4.1. Construcción del corpus

Se cuantificarán las recomendaciones positivas y negativas para evaluar el balance de la variable objetivo.

### 4.5. Resultado esperado

```python

# ============================================================
# STEAM ANALYTICS
# NOTEBOOK 2 - CELDA 2
# CONSTRUCCIÓN Y AUDITORÍA DEL CORPUS
# ============================================================

print("=" * 85)
print("CONSTRUCCIÓN Y AUDITORÍA DEL CORPUS NLP")
print("=" * 85)

# ------------------------------------------------------------
# 1. INTEGRAR RESEÑAS CON DIMENSIONES
# ------------------------------------------------------------

corpus = df_resenas.merge(
    df_usuarios,
    on="user_key",
    how="left",
    validate="many_to_one"
)

corpus = corpus.merge(
    df_idiomas,
    on="language_key",
    how="left",
    validate="many_to_one"
)

print("\n1. INTEGRACIÓN")
print(f"Registros integrados: {len(corpus):,}")

# ------------------------------------------------------------
# 2. NORMALIZACIÓN TÉCNICA INICIAL
# ------------------------------------------------------------

corpus["account_created_year"] = pd.to_numeric(
    corpus["account_created_year"],
    errors="coerce"
)

corpus["language_code"] = (
    corpus["language_code"]
    .astype("string")
    .str.lower()
)

corpus["review_text"] = (
    corpus["review"]
    .fillna("")
    .astype("string")
    .str.strip()
)

# Validar clase objetivo
if not pd.api.types.is_bool_dtype(corpus["voted_up"]):
    corpus["voted_up"] = (
        corpus["voted_up"]
        .astype("string")
        .str.lower()
        .map({"true": True, "false": False})
    )

# ------------------------------------------------------------
# 3. CREAR FILTROS DE SELECCIÓN
# ------------------------------------------------------------

filtro_cohorte = corpus[
    "account_created_year"
].between(
    ANIO_INICIAL,
    ANIO_FINAL
).fillna(False)

filtro_idioma = (
    corpus["language_code"] == IDIOMA_PRINCIPAL
)

filtro_texto = corpus["review_text"].str.len() > 0

filtro_usuario = (
    corpus["user_key"].notna() &
    (corpus["user_key"] > 0)
)

filtro_objetivo = corpus["voted_up"].notna()

filtro_id = corpus["recommendationid"].notna()

# ------------------------------------------------------------
# 4. AUDITAR CADA FILTRO
# ------------------------------------------------------------

auditoria_filtros = pd.DataFrame([
    {
        "criterio": "Reseñas originales",
        "registros": len(corpus)
    },
    {
        "criterio": "Cuentas creadas entre 2020 y 2025",
        "registros": int(filtro_cohorte.sum())
    },
    {
        "criterio": "Reseñas en inglés",
        "registros": int(filtro_idioma.sum())
    },
    {
        "criterio": "Textos no vacíos",
        "registros": int(filtro_texto.sum())
    },
    {
        "criterio": "Usuarios identificados",
        "registros": int(filtro_usuario.sum())
    },
    {
        "criterio": "Recomendación válida",
        "registros": int(filtro_objetivo.sum())
    },
    {
        "criterio": "Identificador presente",
        "registros": int(filtro_id.sum())
    }
])

display(auditoria_filtros)

# ------------------------------------------------------------
# 5. SELECCIONAR CORPUS PRINCIPAL
# ------------------------------------------------------------

seleccion = (
    filtro_cohorte &
    filtro_idioma &
    filtro_texto &
    filtro_usuario &
    filtro_objetivo &
    filtro_id
)

corpus_nlp = corpus.loc[seleccion].copy()

# ------------------------------------------------------------
# 6. ELIMINACIÓN DE DUPLICADOS DE IDENTIFICADOR
# ------------------------------------------------------------

if corpus_nlp["recommendationid"].duplicated().any():
    raise ValueError(
        "Hay recommendationid duplicados en el corpus seleccionado."
    )

# ------------------------------------------------------------
# 7. VARIABLES DEL CORPUS
# ------------------------------------------------------------

COLUMNAS_CORPUS = [
    "recommendationid",
    "user_key",
    "game_key",
    "account_created_year",
    "language_code",
    "timestamp_created_parsed",
    "review_text",
    "voted_up"
]

corpus_nlp = corpus_nlp[
    COLUMNAS_CORPUS
].copy()

# Asegurar tipos
corpus_nlp["account_created_year"] = (
    corpus_nlp["account_created_year"].astype(int)
)

corpus_nlp["voted_up"] = (
    corpus_nlp["voted_up"].astype(bool)
)

# ------------------------------------------------------------
# 8. AUDITORÍA DE TEXTOS
# ------------------------------------------------------------

texto_normalizado = (
    corpus_nlp["review_text"]
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

cantidad_textos = len(corpus_nlp)

textos_unicos = texto_normalizado.nunique()

textos_repetidos = (
    cantidad_textos - textos_unicos
)

# ------------------------------------------------------------
# 9. DISTRIBUCIÓN DE CLASES
# ------------------------------------------------------------

distribucion_clases = (
    corpus_nlp["voted_up"]
    .map({
        True: "Positiva",
        False: "Negativa"
    })
    .value_counts()
    .rename_axis("recomendacion")
    .reset_index(name="cantidad")
)

distribucion_clases["porcentaje"] = (
    distribucion_clases["cantidad"] /
    cantidad_textos * 100
).round(2)

# ------------------------------------------------------------
# 10. DISTRIBUCIÓN POR COHORTE
# ------------------------------------------------------------

distribucion_cohortes = (
    corpus_nlp
    .groupby("account_created_year")
    .agg(
        reseñas=("recommendationid", "count"),
        usuarios=("user_key", "nunique"),
        juegos=("game_key", "nunique"),
        tasa_positiva=("voted_up", "mean")
    )
    .reset_index()
)

distribucion_cohortes["tasa_positiva_pct"] = (
    distribucion_cohortes["tasa_positiva"] * 100
)

# ------------------------------------------------------------
# 11. ESTADÍSTICAS DE LONGITUD
# ------------------------------------------------------------

corpus_nlp["longitud_caracteres"] = (
    corpus_nlp["review_text"].str.len()
)

corpus_nlp["longitud_palabras"] = (
    corpus_nlp["review_text"]
    .str.split()
    .str.len()
)

estadisticas_texto = (
    corpus_nlp[
        ["longitud_caracteres", "longitud_palabras"]
    ]
    .describe(
        percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
    )
    .T
)

# ------------------------------------------------------------
# 12. RESUMEN GENERAL
# ------------------------------------------------------------

resumen_corpus = pd.DataFrame([
    ("Reseñas originales", len(df_resenas)),
    ("Reseñas del corpus seleccionado", cantidad_textos),
    ("Usuarios distintos", corpus_nlp["user_key"].nunique()),
    ("Juegos distintos", corpus_nlp["game_key"].nunique()),
    ("Cohorte inicial", ANIO_INICIAL),
    ("Cohorte final", ANIO_FINAL),
    ("Idioma", IDIOMA_PRINCIPAL),
    ("Textos normalizados únicos", textos_unicos),
    ("Repeticiones textuales excedentes", textos_repetidos),
    ("Identificadores duplicados", corpus_nlp["recommendationid"].duplicated().sum())
], columns=["indicador", "valor"])

# ------------------------------------------------------------
# 13. PRESENTAR RESULTADOS
# ------------------------------------------------------------

print("\n2. RESUMEN DEL CORPUS")
display(resumen_corpus)

print("\n3. DISTRIBUCIÓN DE CLASES")
display(distribucion_clases)

print("\n4. DISTRIBUCIÓN POR COHORTE")
display(distribucion_cohortes.round(3))

print("\n5. LONGITUD DEL TEXTO")
display(estadisticas_texto.round(2))

# ------------------------------------------------------------
# 14. VALIDACIONES FINALES
# ------------------------------------------------------------

if corpus_nlp.empty:
    raise ValueError("El corpus seleccionado está vacío.")

if corpus_nlp["review_text"].str.strip().eq("").any():
    raise ValueError("El corpus contiene textos vacíos.")

if corpus_nlp["voted_up"].isna().any():
    raise ValueError("El corpus contiene objetivos nulos.")

if not corpus_nlp["account_created_year"].between(
    ANIO_INICIAL, ANIO_FINAL
).all():
    raise ValueError("Hay usuarios fuera de las cohortes definidas.")

if set(corpus_nlp["language_code"].unique()) != {"english"}:
    raise ValueError("El corpus contiene idiomas no permitidos.")

# ------------------------------------------------------------
# 15. EXPORTAR INFORMES DE AUDITORÍA
# ------------------------------------------------------------

auditoria_filtros.to_csv(
    REPORTS / "auditoria_filtros_corpus.csv",
    index=False,
    encoding="utf-8-sig"
)

resumen_corpus.to_csv(
    REPORTS / "resumen_corpus_nlp.csv",
    index=False,
    encoding="utf-8-sig"
)

distribucion_clases.to_csv(
    REPORTS / "distribucion_clases_nlp.csv",
    index=False,
    encoding="utf-8-sig"
)

distribucion_cohortes.to_csv(
    REPORTS / "distribucion_cohortes_nlp.csv",
    index=False,
    encoding="utf-8-sig"
)

estadisticas_texto.to_csv(
    REPORTS / "estadisticas_texto_nlp.csv",
    encoding="utf-8-sig"
)

# ------------------------------------------------------------
# 16. RESULTADO FINAL
# ------------------------------------------------------------

print("\n" + "=" * 85)
print("AUDITORÍA DEL CORPUS FINALIZADA")
print("=" * 85)

print(f"Corpus: {len(corpus_nlp):,} reseñas")
print(f"Usuarios: {corpus_nlp['user_key'].nunique():,}")
print(f"Juegos: {corpus_nlp['game_key'].nunique():,}")
print(f"Idioma: {IDIOMA_PRINCIPAL}")
print(f"Cohortes: {ANIO_INICIAL}-{ANIO_FINAL}")

print("\nARCHIVOS DE AUDITORÍA GENERADOS:")
print(REPORTS)

print("\nEl corpus está disponible en memoria como corpus_nlp.")
```


## 5. Análisis exploratorio del contenido textual

### 5.1. Objetivo

Explorar el contenido lingüístico de las reseñas para identificar términos frecuentes, características textuales y posibles diferencias entre recomendaciones positivas y negativas.

### 5.2. Alcance

Se utilizará el corpus en inglés perteneciente a usuarios con cuentas creadas entre 2020 y 2025.

### 5.3. Análisis previstos

- Frecuencia de palabras y expresiones.
- Términos frecuentes en recomendaciones positivas.
- Términos frecuentes en recomendaciones negativas.
- Distribución de la longitud de las reseñas.
- Comparación de características textuales entre clases.
- Auditoría de textos repetidos.
- Distribución de reseñas por usuario y videojuego.

### 5.4. Metodología

Se utilizará una representación inicial mediante frecuencias de términos y n-gramas.

Se excluirán palabras funcionales comunes del inglés mediante una lista de stopwords.

Esta etapa es exploratoria: todavía no se eliminarán palabras del corpus original ni se modificarán los textos que posteriormente utilizaremos para entrenar modelos.

### 5.5. Consideraciones

Las frecuencias observadas no representan por sí mismas importancia predictiva.

Las palabras más frecuentes pueden aparecer en ambas clases. Por ello, su interpretación requerirá considerar el contexto, la distribución de clases y los experimentos posteriores.

Los textos repetidos se investigarán sin eliminarlos automáticamente.

### 5.6. Resultado esperado

- Inventario de términos frecuentes.
- Comparación lingüística entre recomendaciones positivas y negativas.
- Auditoría de duplicados textuales.
- Visualizaciones de longitud y frecuencia.
- Primer diagnóstico del vocabulario del corpus.


```python
# ============================================================
# STEAM ANALYTICS
# NOTEBOOK 2 - CELDA 3
# EXPLORACIÓN DEL CONTENIDO TEXTUAL
# ============================================================

print("=" * 85)
print("EXPLORACIÓN TEXTUAL DEL CORPUS NLP")
print("=" * 85)

# ------------------------------------------------------------
# 0. FUNCIÓN AUXILIAR DE GUARDADO
# ------------------------------------------------------------
def guardar_figura(figura, nombre_archivo):
    ruta_final = VISUALIZATIONS / nombre_archivo
    figura.savefig(ruta_final, bbox_inches="tight", dpi=110)
    print(f"[Visualización Guardada] {ruta_final}")

# ------------------------------------------------------------
# 1. PREPARACIÓN EXPLORATORIA
# ------------------------------------------------------------

texto_eda = corpus_nlp[
    ["recommendationid", "user_key", "game_key",
     "review_text", "voted_up"]
].copy()

texto_eda["texto_normalizado"] = (
    texto_eda["review_text"]
    .fillna("")
    .astype("string")
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

texto_eda["cantidad_palabras"] = (
    texto_eda["texto_normalizado"]
    .str.split()
    .str.len()
)

texto_eda["cantidad_caracteres"] = (
    texto_eda["texto_normalizado"].str.len()
)

texto_eda["clase"] = texto_eda["voted_up"].map({
    True: "Positiva",
    False: "Negativa"
})

# ------------------------------------------------------------
# 2. AUDITORÍA DE TEXTOS REPETIDOS
# ------------------------------------------------------------

duplicados = texto_eda[
    texto_eda["texto_normalizado"].duplicated(keep=False)
].copy()

resumen_duplicados = pd.DataFrame([{
    "total_resenas": len(texto_eda),
    "textos_unicos_normalizados": texto_eda[
        "texto_normalizado"
    ].nunique(),
    "registros_en_grupos_repetidos": len(duplicados),
    "grupos_de_texto_repetidos": duplicados[
        "texto_normalizado"
    ].nunique(),
    "registros_excedentes": int(
        texto_eda["texto_normalizado"].duplicated().sum()
    ),
    "grupos_repetidos_por_varios_usuarios": int(
        duplicados.groupby("texto_normalizado")["user_key"]
        .nunique().gt(1).sum()
    ),
    "grupos_repetidos_con_distintas_recomendaciones": int(
        duplicados.groupby("texto_normalizado")["voted_up"]
        .nunique().gt(1).sum()
    )
}])

print("\n1. AUDITORÍA DE TEXTOS REPETIDOS")
display(resumen_duplicados)

print("\n2. EJEMPLOS DE REPETICIONES")

# No se muestran los textos originales: solamente frecuencias y clases.
ejemplos_duplicados = (
    duplicados.groupby("texto_normalizado")
    .agg(
        repeticiones=("recommendationid", "size"),
        usuarios=("user_key", "nunique"),
        juegos=("game_key", "nunique"),
        clases=("voted_up", "nunique")
    )
    .sort_values("repeticiones", ascending=False)
    .head(15)
    .reset_index(drop=True)
)

display(ejemplos_duplicados)

# ------------------------------------------------------------
# 3. ESTADÍSTICAS DE LONGITUD POR CLASE
# ------------------------------------------------------------

estadisticas_longitud = (
    texto_eda
    .groupby("clase")
    .agg(
        cantidad=("recommendationid", "count"),
        media_palabras=("cantidad_palabras", "mean"),
        mediana_palabras=("cantidad_palabras", "median"),
        media_caracteres=("cantidad_caracteres", "mean"),
        mediana_caracteres=("cantidad_caracteres", "median")
    )
    .reset_index()
)

print("\n3. LONGITUD SEGÚN RECOMENDACIÓN")
display(estadisticas_longitud.round(2))

# ------------------------------------------------------------
# 4. FRECUENCIAS DE TÉRMINOS
# ------------------------------------------------------------

def calcular_frecuencias(textos, etiqueta, ngramas=(1, 2)):

    vectorizador = CountVectorizer(
        stop_words="english",
        lowercase=True,
        ngram_range=ngramas,
        min_df=3,
        max_df=0.90,
        max_features=3000,
        token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z']+\b"
    )

    matriz = vectorizador.fit_transform(textos)

    frecuencias = np.asarray(
        matriz.sum(axis=0)
    ).ravel()

    resultado = pd.DataFrame({
        "termino": vectorizador.get_feature_names_out(),
        "frecuencia": frecuencias
    })

    resultado = resultado.sort_values(
        "frecuencia",
        ascending=False
    ).reset_index(drop=True)

    resultado["clase"] = etiqueta

    return resultado


frecuencias_general = calcular_frecuencias(
    texto_eda["review_text"],
    "Todas"
)

frecuencias_positivas = calcular_frecuencias(
    texto_eda.loc[
        texto_eda["voted_up"],
        "review_text"
    ],
    "Positiva"
)

frecuencias_negativas = calcular_frecuencias(
    texto_eda.loc[
        ~texto_eda["voted_up"],
        "review_text"
    ],
    "Negativa"
)

# ------------------------------------------------------------
# 5. MOSTRAR FRECUENCIAS
# ------------------------------------------------------------

print("\n4. VEINTE TÉRMINOS MÁS FRECUENTES - GENERAL")
display(frecuencias_general.head(20))

print("\n5. VEINTE TÉRMINOS MÁS FRECUENTES - POSITIVAS")
display(frecuencias_positivas.head(20))

print("\n6. VEINTE TÉRMINOS MÁS FRECUENTES - NEGATIVAS")
display(frecuencias_negativas.head(20))

# ------------------------------------------------------------
# 6. VISUALIZACIÓN: LONGITUD DE RESEÑAS
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(12, 7))

percentil_99 = texto_eda["cantidad_palabras"].quantile(0.99)

for clase, color in [
    ("Positiva", "#24966D"),
    ("Negativa", "#D97757")
]:

    valores = texto_eda.loc[
        texto_eda["clase"] == clase,
        "cantidad_palabras"
    ]

    valores = valores[valores <= percentil_99]

    ax.hist(
        valores,
        bins=45,
        alpha=0.55,
        label=clase,
        color=color
    )

ax.set_title(
    "Distribución de longitud de reseñas por recomendación",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Cantidad de palabras")
ax.set_ylabel("Frecuencia")
ax.legend()
ax.grid(alpha=0.2)

guardar_figura(
    fig,
    "01_longitud_textual_por_clase.png"
)

# ------------------------------------------------------------
# 7. VISUALIZACIÓN: TÉRMINOS FRECUENTES
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(16, 8))

for ax, dataframe, titulo, color in [
    (
        axes[0],
        frecuencias_positivas.head(15).sort_values("frecuencia"),
        "Términos frecuentes - Positivas",
        "#24966D"
    ),
    (
        axes[1],
        frecuencias_negativas.head(15).sort_values("frecuencia"),
        "Términos frecuentes - Negativas",
        "#D97757"
    )
]:

    ax.barh(
        dataframe["termino"],
        dataframe["frecuencia"],
        color=color
    )

    ax.set_title(
        titulo,
        fontsize=12,
        fontweight="bold"
    )

    ax.set_xlabel("Frecuencia")
    ax.set_ylabel("Término")
    ax.tick_params(axis="y", labelsize=9)
    ax.grid(axis="x", alpha=0.2)

fig.suptitle(
    "Comparación de vocabulario por tipo de recomendación",
    fontsize=15,
    fontweight="bold"
)

guardar_figura(
    fig,
    "02_frecuencias_por_recomendacion.png"
)

# ------------------------------------------------------------
# 8. VISUALIZACIÓN: DISTRIBUCIÓN POR CLASE
# ------------------------------------------------------------

conteos = (
    texto_eda["clase"]
    .value_counts()
    .reindex(["Positiva", "Negativa"])
)

fig, ax = plt.subplots(figsize=(9, 5))

barras = ax.bar(
    conteos.index,
    conteos.values,
    color=["#24966D", "#D97757"],
    width=0.55
)

ax.bar_label(
    barras,
    labels=[
        f"{v:,} ({v / len(texto_eda) * 100:.2f}%)"
        for v in conteos
    ],
    padding=5
)

ax.set_title(
    "Distribución de clases del corpus de NLP",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Recomendación")
ax.set_ylabel("Cantidad de reseñas")
ax.grid(axis="y", alpha=0.2)

guardar_figura(
    fig,
    "03_balance_clases_corpus.png"
)

# ------------------------------------------------------------
# 9. EXPORTAR RESULTADOS
# ------------------------------------------------------------

EXPORTACIONES = {
    "auditoria_duplicados_textuales.csv": resumen_duplicados,
    "longitud_textual_por_clase.csv": estadisticas_longitud,
    "frecuencias_terminos_general.csv": frecuencias_general,
    "frecuencias_terminos_positivas.csv": frecuencias_positivas,
    "frecuencias_terminos_negativas.csv": frecuencias_negativas
}

print("\n7. EXPORTACIÓN DE RESULTADOS")

for archivo, dataframe in EXPORTACIONES.items():

    dataframe.to_csv(
        REPORTS / archivo,
        index=False,
        encoding="utf-8-sig"
    )

    print(f"[OK] {archivo}")

print("\n" + "=" * 85)
print("EXPLORACIÓN TEXTUAL FINALIZADA")
print("=" * 85)
```


## 6. Limpieza y normalización textual

### 6.1. Objetivo

Preparar el contenido textual de las reseñas mediante operaciones de normalización que faciliten su posterior procesamiento y representación numérica.

### 6.2. Transformaciones

Se realizarán las siguientes operaciones:

- Normalización Unicode.
- Conversión a minúsculas.
- Decodificación de entidades HTML.
- Eliminación de etiquetas HTML.
- Sustitución de direcciones URL.
- Eliminación de caracteres de control.
- Normalización de espacios en blanco.

### 6.3. Conservación de información lingüística

Se conservarán las negaciones, contracciones y términos que puedan transmitir información relevante sobre las opiniones de los usuarios.

No se eliminarán inicialmente todas las palabras funcionales ni se aplicarán técnicas de stemming o lematización.

Estas operaciones se evaluarán posteriormente para determinar si mejoran el rendimiento de los modelos.

### 6.4. Conservación del corpus original

Se mantendrá la columna `review_text` sin modificaciones.

Las transformaciones se almacenarán en una nueva columna denominada `review_clean`.

De esta manera será posible comparar el texto original con el procesado y reproducir diferentes estrategias de preprocesamiento.

### 6.5. Controles de calidad

Se comprobarán:

- Textos vacíos después de la normalización.
- Longitud original y procesada.
- Cantidad de textos repetidos.
- Grupos de textos asociados a diferentes recomendaciones.
- Conservación de identificadores y variables objetivo.

### 6.6. Resultado esperado

Obtener una versión normalizada del corpus para posteriores experimentos de vectorización TF-IDF, extracción de características y clasificación de reseñas.


```python

# ============================================================
# LIMPIEZA Y NORMALIZACIÓN TEXTUAL
# ============================================================

print("=" * 85)
print("LIMPIEZA Y NORMALIZACIÓN TEXTUAL")
print("=" * 85)

# ------------------------------------------------------------
# 1. CONSERVAR CORPUS ORIGINAL
# ------------------------------------------------------------

corpus_procesado = corpus_nlp.copy()

# ------------------------------------------------------------
# 2. FUNCIÓN DE NORMALIZACIÓN
# ------------------------------------------------------------

def normalizar_texto(texto):

    if pd.isna(texto):
        return ""

    texto = str(texto)

    # Decodificar entidades HTML
    texto = html.unescape(texto)

    # Normalizar caracteres Unicode
    texto = unicodedata.normalize("NFKC", texto)

    # Eliminar etiquetas HTML
    texto = re.sub(
        r"<[^>]*>",
        " ",
        texto
    )

    # Sustituir URLs por un token
    texto = re.sub(
        r"https?://\S+|www\.\S+",
        " URLTOKEN ",
        texto,
        flags=re.IGNORECASE
    )

    # Eliminar caracteres de control
    texto = re.sub(
        r"[\x00-\x08\x0B\x0C\x0E-\x1F\x7F]",
        " ",
        texto
    )

    # Convertir a minúsculas
    texto = texto.lower()

    # Normalizar espacios
    texto = re.sub(
        r"\s+",
        " ",
        texto
    ).strip()

    return texto


# ------------------------------------------------------------
# 3. GENERAR TEXTO NORMALIZADO
# ------------------------------------------------------------

corpus_procesado["review_clean"] = (
    corpus_procesado["review_text"]
    .apply(normalizar_texto)
)

# ------------------------------------------------------------
# 4. VALIDAR TEXTOS
# ------------------------------------------------------------

total_original = len(corpus_procesado)

textos_vacios_originales = int(
    corpus_procesado["review_text"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

textos_vacios_procesados = int(
    corpus_procesado["review_clean"].eq("").sum()
)

textos_modificados = int(
    (
        corpus_procesado["review_text"].astype(str) !=
        corpus_procesado["review_clean"]
    ).sum()
)

# ------------------------------------------------------------
# 5. COMPARAR LONGITUDES
# ------------------------------------------------------------

corpus_procesado["caracteres_originales"] = (
    corpus_procesado["review_text"].str.len()
)

corpus_procesado["caracteres_procesados"] = (
    corpus_procesado["review_clean"].str.len()
)

# ------------------------------------------------------------
# 6. AUDITORÍA DE REPETICIONES
# ------------------------------------------------------------

textos_validos = corpus_procesado[
    corpus_procesado["review_clean"] != ""
].copy()

grupos_texto = (
    textos_validos
    .groupby("review_clean")
    .agg(
        registros=("recommendationid", "size"),
        usuarios=("user_key", "nunique"),
        videojuegos=("game_key", "nunique"),
        recomendaciones=("voted_up", "nunique")
    )
    .reset_index()
)

grupos_repetidos = grupos_texto[
    grupos_texto["registros"] > 1
].copy()

grupos_conflictivos = grupos_repetidos[
    grupos_repetidos["recomendaciones"] > 1
].copy()

# ------------------------------------------------------------
# 7. RESUMEN DE PROCESAMIENTO
# ------------------------------------------------------------

auditoria_preprocesamiento = pd.DataFrame([
    ("Registros originales", total_original),
    ("Registros con texto vacío original", textos_vacios_originales),
    ("Textos modificados", textos_modificados),
    ("Textos vacíos después de normalizar", textos_vacios_procesados),
    ("Textos válidos procesados", len(textos_validos)),
    ("Textos normalizados únicos", textos_validos["review_clean"].nunique()),
    ("Grupos repetidos", len(grupos_repetidos)),
    ("Registros en grupos repetidos", int(grupos_repetidos["registros"].sum())),
    ("Grupos con etiquetas diferentes", len(grupos_conflictivos)),
    ("Registros excedentes", int(
        len(textos_validos) -
        textos_validos["review_clean"].nunique()
    ))
], columns=["indicador", "valor"])

print("\n1. AUDITORÍA GENERAL")
display(auditoria_preprocesamiento)

# ------------------------------------------------------------
# 8. COMPARACIÓN DE EJEMPLOS
# ------------------------------------------------------------

print("\n2. EJEMPLOS DE TEXTO ORIGINAL Y NORMALIZADO")

display(
    corpus_procesado[
        ["review_text", "review_clean", "voted_up"]
    ].head(10)
)

# ------------------------------------------------------------
# 9. AUDITORÍA DE ETIQUETAS CONFLICTIVAS
# ------------------------------------------------------------

print("\n3. GRUPOS DE TEXTOS CON RECOMENDACIONES DIFERENTES")

print(
    f"Grupos conflictivos: {len(grupos_conflictivos)}"
)

print(
    "No se eliminarán automáticamente estos registros."
)

# ------------------------------------------------------------
# 10. VALIDACIONES
# ------------------------------------------------------------

if len(corpus_procesado) != len(corpus_nlp):
    raise ValueError("Se modificó la cantidad de registros.")

if corpus_procesado["recommendationid"].duplicated().any():
    raise ValueError("Se detectaron recommendationid duplicados.")

if not corpus_procesado["recommendationid"].equals(
    corpus_nlp["recommendationid"]
):
    raise ValueError("Se alteró el orden o identidad de los registros.")

if not corpus_procesado["voted_up"].equals(
    corpus_nlp["voted_up"]
):
    raise ValueError("Se modificó la variable objetivo.")

# ------------------------------------------------------------
# 11. EXPORTAR INFORMES
# ------------------------------------------------------------

auditoria_preprocesamiento.to_csv(
    REPORTS / "auditoria_preprocesamiento_textual.csv",
    index=False,
    encoding="utf-8-sig"
)

grupos_conflictivos.drop(
    columns=["review_clean"]
).to_csv(
    REPORTS / "grupos_textuales_etiquetas_conflictivas.csv",
    index=False,
    encoding="utf-8-sig"
)

# ------------------------------------------------------------
# 12. RESULTADO FINAL
# ------------------------------------------------------------

print("\n" + "=" * 85)
print("NORMALIZACIÓN TEXTUAL FINALIZADA")
print("=" * 85)

print(f"Registros conservados: {len(corpus_procesado):,}")
print(f"Textos válidos: {len(textos_validos):,}")
print(f"Textos vacíos posteriores: {textos_vacios_procesados}")
print(f"Grupos de etiquetas conflictivas: {len(grupos_conflictivos)}")

print("\nVariable de texto procesado: review_clean")
print("Corpus original conservado: review_text")

print(f"\nInformes guardados en: {REPORTS}")
```


## 7. Preparación del conjunto de datos para Machine Learning

### 7.1. Objetivo

Preparar una partición de entrenamiento y prueba que permita evaluar los futuros modelos de clasificación textual evitando fugas de información entre usuarios y textos repetidos.

### 7.2. Tratamiento de textos ambiguos

Durante la auditoría se identificaron tres grupos de textos normalizados asociados a recomendaciones diferentes.

Estos registros se excluirán únicamente del conjunto principal de entrenamiento supervisado, porque un mismo texto presenta etiquetas contradictorias.

Los registros originales permanecerán conservados en el corpus procesado y en los informes de auditoría.

### 7.3. Control de duplicados y usuarios

Se mantendrán los textos repetidos que presenten etiquetas consistentes.

Sin embargo, las reseñas de un mismo usuario y los textos idénticos normalizados deberán permanecer en una misma partición.

Se utilizará una estrategia de agrupamiento mediante componentes conectados y separación estratificada por grupos.

### 7.4. Separación de datos

Se implementará una partición aproximada del 80 % para entrenamiento y del 20 % para prueba.

Se utilizará StratifiedGroupKFold con cinco particiones y una semilla fija para favorecer la reproducibilidad y conservar la distribución de clases.

### 7.5. Resultado esperado

- Conjunto de entrenamiento.
- Conjunto de prueba.
- Distribución de recomendaciones por partición.
- Control de usuarios compartidos.
- Control de textos duplicados entre particiones.
- Registro reproducible de la separación.

La vectorización TF-IDF se ajustará exclusivamente con el conjunto de entrenamiento.


```python

# ============================================================
# STEAM ANALYTICS
# NOTEBOOK 2 - CELDA 5
# PREPARACIÓN DE DATOS PARA MACHINE LEARNING
# ============================================================

print("=" * 85)
print("PREPARACIÓN DEL CONJUNTO DE MODELADO")
print("=" * 85)

# ------------------------------------------------------------
# 1. IDENTIFICAR TEXTOS CON ETIQUETAS CONTRADICTORIAS
# ------------------------------------------------------------

etiquetas_por_texto = (
    corpus_procesado
    .groupby("review_clean")["voted_up"]
    .nunique()
)

textos_conflictivos = etiquetas_por_texto[
    etiquetas_por_texto > 1
].index

mask_conflictivos = (
    corpus_procesado["review_clean"].isin(
        textos_conflictivos
    )
)

registros_conflictivos = int(mask_conflictivos.sum())

# Excluir solamente del corpus supervisado principal.
# El corpus original continúa intacto.
corpus_modelo = (
    corpus_procesado.loc[~mask_conflictivos]
    .copy()
    .reset_index(drop=True)
)

if corpus_modelo.empty:
    raise ValueError("El conjunto de modelado está vacío.")

if corpus_modelo["voted_up"].nunique() != 2:
    raise ValueError("El corpus debe contener ambas clases.")

print("\n1. TRATAMIENTO DE ETIQUETAS")
print(f"Grupos con etiquetas contradictorias: {len(textos_conflictivos)}")
print(f"Registros excluidos del modelo principal: {registros_conflictivos}")
print(f"Registros conservados: {len(corpus_modelo):,}")

# ------------------------------------------------------------
# 2. AGRUPAMIENTO POR USUARIO Y TEXTO
# ------------------------------------------------------------

# Se utiliza Union-Find para encontrar componentes conectados.
# Dos filas quedan en el mismo grupo cuando:
# - Comparten usuario.
# - Comparten texto normalizado.
# - O existe una cadena de conexiones entre ambos criterios.

n = len(corpus_modelo)

parent = list(range(n))
rank = [0] * n

def find_root(x):

    while parent[x] != x:
        parent[x] = parent[parent[x]]
        x = parent[x]

    return x


def union(x, y):

    root_x = find_root(x)
    root_y = find_root(y)

    if root_x == root_y:
        return

    if rank[root_x] < rank[root_y]:
        root_x, root_y = root_y, root_x

    parent[root_y] = root_x

    if rank[root_x] == rank[root_y]:
        rank[root_x] += 1


referencia_usuario = {}
referencia_texto = {}

for i, (usuario, texto) in enumerate(
    zip(
        corpus_modelo["user_key"],
        corpus_modelo["review_clean"]
    )
):

    usuario = int(usuario)
    texto = str(texto)

    # Conectar reseñas del mismo usuario.
    if usuario in referencia_usuario:
        union(i, referencia_usuario[usuario])
    else:
        referencia_usuario[usuario] = i

    # Conectar textos idénticos normalizados.
    if texto in referencia_texto:
        union(i, referencia_texto[texto])
    else:
        referencia_texto[texto] = i


# Identificadores de componentes conectados.
raices = [
    find_root(i)
    for i in range(n)
]

grupos, componentes = pd.factorize(raices)

corpus_modelo["grupo_particion"] = grupos.astype("int64")

cantidad_grupos = len(componentes)

print("\n2. AGRUPAMIENTO")
print(f"Registros: {n:,}")
print(f"Usuarios únicos: {corpus_modelo['user_key'].nunique():,}")
print(f"Textos únicos: {corpus_modelo['review_clean'].nunique():,}")
print(f"Componentes conectados: {cantidad_grupos:,}")

if cantidad_grupos < 5:
    raise ValueError(
        "No existen suficientes grupos para una separación en cinco folds."
    )

# ------------------------------------------------------------
# 3. SEPARACIÓN ESTRATIFICADA POR GRUPOS
# ------------------------------------------------------------

X_indices = np.zeros(n)

y = corpus_modelo["voted_up"].astype(int)

groups = corpus_modelo["grupo_particion"]

cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

splits = list(
    cv.split(X_indices, y, groups)
)

train_idx, test_idx = splits[0]

X_train_text = (
    corpus_modelo.iloc[train_idx]["review_clean"]
    .reset_index(drop=True)
)

X_test_text = (
    corpus_modelo.iloc[test_idx]["review_clean"]
    .reset_index(drop=True)
)

y_train = (
    corpus_modelo.iloc[train_idx]["voted_up"]
    .astype(int)
    .reset_index(drop=True)
)

y_test = (
    corpus_modelo.iloc[test_idx]["voted_up"]
    .astype(int)
    .reset_index(drop=True)
)

train_data = corpus_modelo.iloc[train_idx].copy()
test_data = corpus_modelo.iloc[test_idx].copy()

# ------------------------------------------------------------
# 4. CONTROL DE FUGA DE INFORMACIÓN
# ------------------------------------------------------------

usuarios_train = set(train_data["user_key"])
usuarios_test = set(test_data["user_key"])

textos_train = set(train_data["review_clean"])
textos_test = set(test_data["review_clean"])

grupos_train = set(train_data["grupo_particion"])
grupos_test = set(test_data["grupo_particion"])

usuarios_compartidos = usuarios_train & usuarios_test
textos_compartidos = textos_train & textos_test
grupos_compartidos = grupos_train & grupos_test

if usuarios_compartidos:
    raise ValueError("Se detectaron usuarios compartidos.")

if textos_compartidos:
    raise ValueError("Se detectaron textos compartidos.")

if grupos_compartidos:
    raise ValueError("Se detectaron grupos compartidos.")

print("\n3. CONTROL DE FUGA DE INFORMACIÓN")
print("Usuarios compartidos:", len(usuarios_compartidos))
print("Textos compartidos:", len(textos_compartidos))
print("Grupos compartidos:", len(grupos_compartidos))

# ------------------------------------------------------------
# 5. DISTRIBUCIÓN DE CLASES
# ------------------------------------------------------------

def resumen_clases(y_valores, nombre):

    conteo = pd.Series(y_valores).value_counts()

    total = len(y_valores)

    return {
        "particion": nombre,
        "registros": total,
        "positivas": int(conteo.get(1, 0)),
        "negativas": int(conteo.get(0, 0)),
        "porcentaje_positivas": round(
            conteo.get(1, 0) / total * 100, 2
        ),
        "porcentaje_negativas": round(
            conteo.get(0, 0) / total * 100, 2
        )
    }


balance_particiones = pd.DataFrame([
    resumen_clases(y_train, "Entrenamiento"),
    resumen_clases(y_test, "Prueba")
])

print("\n4. DISTRIBUCIÓN DE CLASES")
display(balance_particiones)

# ------------------------------------------------------------
# 6. TAMAÑO DE LAS PARTICIONES
# ------------------------------------------------------------

total_modelo = len(corpus_modelo)

print("\n5. TAMAÑO DE LAS PARTICIONES")

print(
    f"Entrenamiento: {len(train_idx):,} "
    f"({len(train_idx) / total_modelo * 100:.2f}%)"
)

print(
    f"Prueba: {len(test_idx):,} "
    f"({len(test_idx) / total_modelo * 100:.2f}%)"
)

# ------------------------------------------------------------
# 7. VALIDACIÓN FINAL
# ------------------------------------------------------------

if len(train_idx) + len(test_idx) != total_modelo:
    raise ValueError("No se conservaron todos los registros.")

if set(train_data["recommendationid"]) & set(
    test_data["recommendationid"]
):
    raise ValueError("Hay recommendationid compartidos.")

if y_train.nunique() != 2 or y_test.nunique() != 2:
    raise ValueError(
        "Alguna partición no contiene ambas clases."
    )

print("\n6. VALIDACIONES")
print("[OK] Usuarios separados.")
print("[OK] Textos idénticos separados.")
print("[OK] Componentes conectados separados.")
print("[OK] Ambas clases presentes.")
print("[OK] Identificadores sin solapamiento.")

# ------------------------------------------------------------
# 8. EXPORTAR MANIFIESTO REPRODUCIBLE
# ------------------------------------------------------------

manifiesto = corpus_modelo[
    [
        "recommendationid",
        "grupo_particion"
    ]
].copy()

manifiesto["particion"] = "entrenamiento"

manifiesto.loc[test_idx, "particion"] = "prueba"

manifiesto.to_csv(
    REPORTS / "manifiesto_particion_ml.csv",
    index=False,
    encoding="utf-8-sig"
)

balance_particiones.to_csv(
    REPORTS / "balance_particiones_ml.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n7. ARCHIVOS GENERADOS")
print(REPORTS / "manifiesto_particion_ml.csv")
print(REPORTS / "balance_particiones_ml.csv")

print("\n" + "=" * 85)
print("PARTICIÓN PARA MACHINE LEARNING COMPLETADA")
print("=" * 85)
```


## 8. Representación textual mediante TF-IDF (Frecuencia de término – Frecuencia inversa de documento)

### 8.1. Objetivo

Transformar las reseñas normalizadas en representaciones numéricas que puedan utilizarse en los algoritmos de Machine Learning.

### 8.2. Fundamento

Se utilizará TF-IDF para asignar ponderaciones a términos y expresiones según su relevancia dentro de cada reseña y su distribución en el conjunto de entrenamiento.

### 8.3. Configuración

Se implementará una representación que contemple:

- Unigramas: términos individuales.
- Bigramas: combinaciones consecutivas de dos términos.
- Frecuencia mínima de aparición de dos documentos.
- Límite máximo de 30.000 características.
- Ponderación mediante TF-IDF.
- Normalización L2.
- Ponderación logarítmica de frecuencia.

Se conservarán inicialmente las palabras funcionales, especialmente las negaciones, debido a su importancia para interpretar opiniones.

### 8.4. Prevención de fuga de información

El vectorizador se ajustará exclusivamente sobre los textos de entrenamiento.

El conjunto de prueba solamente será transformado utilizando el vocabulario y las ponderaciones aprendidas durante el entrenamiento.

### 8.5. Resultado esperado

- Matriz TF-IDF de entrenamiento.
- Matriz TF-IDF de prueba.
- Vocabulario de características.
- Diagnóstico de cobertura y dispersión.
- Informe de dimensionalidad.

Estas representaciones serán utilizadas en los experimentos posteriores de clasificación textual.


```python

# ============================================================
# STEAM ANALYTICS
# NOTEBOOK 2 - CELDA 6
# REPRESENTACIÓN TEXTUAL TF-IDF
# ============================================================

print("=" * 85)
print("REPRESENTACIÓN TEXTUAL MEDIANTE TF-IDF")
print("=" * 85)

# ------------------------------------------------------------
# 1. VALIDACIONES INICIALES
# ------------------------------------------------------------

if len(X_train_text) != len(y_train):
    raise ValueError("Entrenamiento: textos y etiquetas incompatibles.")

if len(X_test_text) != len(y_test):
    raise ValueError("Prueba: textos y etiquetas incompatibles.")

if X_train_text.isna().any() or X_test_text.isna().any():
    raise ValueError("Existen textos nulos.")

if X_train_text.str.strip().eq("").any():
    raise ValueError("Hay textos vacíos en entrenamiento.")

if X_test_text.str.strip().eq("").any():
    raise ValueError("Hay textos vacíos en prueba.")

# ------------------------------------------------------------
# 2. CONFIGURACIÓN DEL VECTORIZADOR
# ------------------------------------------------------------

CONFIG_TFIDF = {
    "ngram_range": (1, 2),
    "min_df": 2,
    "max_df": 0.95,
    "max_features": 30000,
    "sublinear_tf": True,
    "norm": "l2",
    "strip_accents": "unicode",
    "lowercase": False,
    "stop_words": None,
    "dtype": np.float32,
    "token_pattern": r"(?u)\b[a-zA-Z][a-zA-Z']+\b"
}

vectorizador_tfidf = TfidfVectorizer(
    **CONFIG_TFIDF
)

# ------------------------------------------------------------
# 3. AJUSTAR SOLAMENTE CON ENTRENAMIENTO
# ------------------------------------------------------------

X_train_tfidf = vectorizador_tfidf.fit_transform(
    X_train_text.astype(str)
)

# El conjunto de prueba no participa en el ajuste.
X_test_tfidf = vectorizador_tfidf.transform(
    X_test_text.astype(str)
)

# ------------------------------------------------------------
# 4. DIMENSIONALIDAD
# ------------------------------------------------------------

n_features = len(
    vectorizador_tfidf.get_feature_names_out()
)

nnz_train = X_train_tfidf.nnz
nnz_test = X_test_tfidf.nnz

densidad_train = (
    nnz_train /
    (X_train_tfidf.shape[0] * X_train_tfidf.shape[1])
) * 100

densidad_test = (
    nnz_test /
    (X_test_tfidf.shape[0] * X_test_tfidf.shape[1])
) * 100

filas_sin_caracteristicas_train = int(
    np.diff(X_train_tfidf.indptr).eq(0).sum()
) if isinstance(np.diff(X_train_tfidf.indptr), pd.Series) else int(
    (np.diff(X_train_tfidf.indptr) == 0).sum()
)

filas_sin_caracteristicas_test = int(
    (np.diff(X_test_tfidf.indptr) == 0).sum()
)

print("\n1. DIMENSIONES DE LAS MATRICES")

print(f"Características: {n_features:,}")

print(
    f"Entrenamiento: "
    f"{X_train_tfidf.shape[0]:,} x "
    f"{X_train_tfidf.shape[1]:,}"
)

print(
    f"Prueba: "
    f"{X_test_tfidf.shape[0]:,} x "
    f"{X_test_tfidf.shape[1]:,}"
)

# ------------------------------------------------------------
# 5. DISPERSIÓN DE LAS MATRICES
# ------------------------------------------------------------

print("\n2. DISPERSIÓN")

print(f"Densidad entrenamiento: {densidad_train:.4f}%")
print(f"Densidad prueba: {densidad_test:.4f}%")

print(
    "Filas sin características - entrenamiento:",
    filas_sin_caracteristicas_train
)

print(
    "Filas sin características - prueba:",
    filas_sin_caracteristicas_test
)

# ------------------------------------------------------------
# 6. COBERTURA DEL VOCABULARIO EN PRUEBA
# ------------------------------------------------------------

analizador = vectorizador_tfidf.build_analyzer()

vocabulario = set(
    vectorizador_tfidf.vocabulary_.keys()
)

tokens_test = [
    termino
    for texto in X_test_text.astype(str)
    for termino in analizador(texto)
]

tokens_cubiertos = sum(
    termino in vocabulario
    for termino in tokens_test
)

cobertura_vocabulario = (
    tokens_cubiertos / len(tokens_test) * 100
    if tokens_test else 0
)

print("\n3. COBERTURA DE VOCABULARIO")
print(f"Tokens analizados: {len(tokens_test):,}")
print(f"Tokens dentro del vocabulario: {tokens_cubiertos:,}")
print(f"Cobertura: {cobertura_vocabulario:.2f}%")

# ------------------------------------------------------------
# 7. INSPECCIÓN DEL VOCABULARIO
# ------------------------------------------------------------

terminos = vectorizador_tfidf.get_feature_names_out()

vocabulario_df = pd.DataFrame({
    "termino": terminos,
    "idf": vectorizador_tfidf.idf_
}).sort_values(
    "idf",
    ascending=True
)

print("\n4. TÉRMINOS CON MENOR IDF")
display(vocabulario_df.head(20))

print("\n5. TÉRMINOS CON MAYOR IDF")
display(vocabulario_df.tail(20))

# ------------------------------------------------------------
# 8. EJEMPLO DE REPRESENTACIÓN
# ------------------------------------------------------------

print("\n6. EJEMPLO DE TEXTO Y REPRESENTACIÓN")

if len(X_train_text) > 0:

    print("Texto original:")
    print(X_train_text.iloc[0][:500])

    fila = X_train_tfidf.getrow(0)

    print("\nCaracterísticas no nulas:")
    print(fila.nnz)

    ejemplo = pd.DataFrame({
        "termino": terminos[fila.indices],
        "peso_tfidf": fila.data
    }).sort_values(
        "peso_tfidf",
        ascending=False
    )

    display(ejemplo.head(15))

# ------------------------------------------------------------
# 9. DIAGNÓSTICO
# ------------------------------------------------------------

diagnostico_tfidf = pd.DataFrame([
    {
        "particion": "Entrenamiento",
        "documentos": X_train_tfidf.shape[0],
        "caracteristicas": X_train_tfidf.shape[1],
        "valores_no_cero": nnz_train,
        "densidad_pct": densidad_train,
        "filas_sin_caracteristicas": filas_sin_caracteristicas_train
    },
    {
        "particion": "Prueba",
        "documentos": X_test_tfidf.shape[0],
        "caracteristicas": X_test_tfidf.shape[1],
        "valores_no_cero": nnz_test,
        "densidad_pct": densidad_test,
        "filas_sin_caracteristicas": filas_sin_caracteristicas_test
    }
])

display(diagnostico_tfidf)

# ------------------------------------------------------------
# 10. EXPORTACIÓN DE INFORMES
# ------------------------------------------------------------

diagnostico_tfidf.to_csv(
    REPORTS / "diagnostico_tfidf.csv",
    index=False,
    encoding="utf-8-sig"
)

vocabulario_df.to_csv(
    REPORTS / "vocabulario_tfidf.csv",
    index=False,
    encoding="utf-8-sig"
)

configuracion_tfidf = pd.DataFrame(
    list(CONFIG_TFIDF.items()),
    columns=["parametro", "valor"]
)

configuracion_tfidf.to_csv(
    REPORTS / "configuracion_tfidf.csv",
    index=False,
    encoding="utf-8-sig"
)

# ------------------------------------------------------------
# 11. VALIDACIÓN FINAL
# ------------------------------------------------------------

if X_train_tfidf.shape[1] != X_test_tfidf.shape[1]:
    raise ValueError("Las dimensiones de las matrices no coinciden.")

if X_train_tfidf.shape[1] == 0:
    raise ValueError("El vocabulario TF-IDF está vacío.")

if not np.isfinite(X_train_tfidf.data).all():
    raise ValueError("Hay valores no finitos en entrenamiento.")

if not np.isfinite(X_test_tfidf.data).all():
    raise ValueError("Hay valores no finitos en prueba.")

print("\n" + "=" * 85)
print("TF-IDF FINALIZADO CORRECTAMENTE")
print("=" * 85)

print(f"Vocabulario: {n_features:,} características")
print(f"Cobertura en prueba: {cobertura_vocabulario:.2f}%")
print(f"Informes guardados en: {REPORTS}")
```


## 8.6. Auditoría de documentos sin representación TF-IDF

### Objetivo

Identificar las reseñas que no generaron características durante la vectorización TF-IDF.

### Procedimiento

Se examinarán las particiones de entrenamiento y prueba para determinar:

- Cantidad de documentos sin tokens reconocidos.
- Cantidad de documentos cuyos términos fueron excluidos del vocabulario.
- Características de los textos afectados.
- Distribución de recomendaciones de estos registros.

### Criterio de tratamiento

No se eliminarán automáticamente estas reseñas.

Se evaluará si resulta necesario modificar el tokenizador o los parámetros de vectorización y se documentará cualquier decisión.

### Resultado esperado

Un diagnóstico de cobertura textual que permita comprobar la calidad de la representación TF-IDF antes del entrenamiento de modelos.


```python

# ============================================================
# STEAM ANALYTICS
# NOTEBOOK 2 - CELDA 7
# AUDITORÍA DE FILAS SIN CARACTERÍSTICAS TF-IDF
# ============================================================

print("=" * 85)
print("AUDITORÍA DE DOCUMENTOS SIN CARACTERÍSTICAS")
print("=" * 85)

vocabulario = vectorizador_tfidf.vocabulary_
analizador = vectorizador_tfidf.build_analyzer()

def diagnosticar_documentos(X, dataframe, particion):

    mascara_vacia = np.diff(X.indptr) == 0

    dataframe = dataframe.reset_index(drop=True)

    if len(dataframe) != X.shape[0]:
        raise ValueError(
            f"Los registros de {particion} no coinciden con TF-IDF."
        )

    filas = []

    for indice in np.flatnonzero(mascara_vacia):

        texto = str(dataframe.loc[indice, "review_clean"])

        tokens = analizador(texto)

        reconocidos = [
            token for token in tokens
            if token in vocabulario
        ]

        if not tokens:
            motivo = "Sin tokens reconocidos por el tokenizador"
        elif not reconocidos:
            motivo = "Todos los términos quedaron fuera del vocabulario"
        else:
            motivo = "Revisar transformación TF-IDF"

        filas.append({
            "particion": particion,
            "indice": int(indice),
            "recommendationid": dataframe.loc[
                indice, "recommendationid"
            ],
            "game_key": dataframe.loc[indice, "game_key"],
            "voted_up": bool(dataframe.loc[indice, "voted_up"]),
            "longitud": len(texto),
            "cantidad_tokens": len(tokens),
            "tokens_reconocidos": len(reconocidos),
            "motivo": motivo,
            "ejemplo_texto": texto[:200]
        })

    return pd.DataFrame(filas)


# ------------------------------------------------------------
# 1. DIAGNOSTICAR ENTRENAMIENTO Y PRUEBA
# ------------------------------------------------------------

train_alineado = train_data.reset_index(drop=True)
test_alineado = test_data.reset_index(drop=True)

sin_caracteristicas_train = diagnosticar_documentos(
    X_train_tfidf,
    train_alineado,
    "Entrenamiento"
)

sin_caracteristicas_test = diagnosticar_documentos(
    X_test_tfidf,
    test_alineado,
    "Prueba"
)

diagnostico_vacios = pd.concat(
    [
        sin_caracteristicas_train,
        sin_caracteristicas_test
    ],
    ignore_index=True
)

# ------------------------------------------------------------
# 2. RESUMEN
# ------------------------------------------------------------

print("\n1. RESUMEN DE DOCUMENTOS SIN CARACTERÍSTICAS")

resumen_vacios = (
    diagnostico_vacios
    .groupby(["particion", "motivo"])
    .size()
    .rename("cantidad")
    .reset_index()
)

display(resumen_vacios)

print("\nTotal de documentos sin características:")
print(f"{len(diagnostico_vacios):,}")

# ------------------------------------------------------------
# 3. DISTRIBUCIÓN POR RECOMENDACIÓN
# ------------------------------------------------------------

print("\n2. DISTRIBUCIÓN DE CLASES")

if not diagnostico_vacios.empty:

    balance_vacios = (
        diagnostico_vacios
        .groupby(["particion", "voted_up"])
        .size()
        .rename("cantidad")
        .reset_index()
    )

    balance_vacios["recomendacion"] = (
        balance_vacios["voted_up"]
        .map({True: "Positiva", False: "Negativa"})
    )

    display(balance_vacios)

# ------------------------------------------------------------
# 4. EJEMPLOS REPRESENTATIVOS
# ------------------------------------------------------------

print("\n3. EJEMPLOS DE TEXTOS AFECTADOS")

if not diagnostico_vacios.empty:

    display(
        diagnostico_vacios[
            [
                "particion",
                "longitud",
                "cantidad_tokens",
                "tokens_reconocidos",
                "motivo",
                "ejemplo_texto"
            ]
        ].head(20)
    )

else:
    print("No hay documentos sin características.")

# ------------------------------------------------------------
# 5. EXPORTAR DIAGNÓSTICO
# ------------------------------------------------------------

ruta_diagnostico = (
    REPORTS / "diagnostico_documentos_sin_tfidf.csv"
)

diagnostico_vacios.to_csv(
    ruta_diagnostico,
    index=False,
    encoding="utf-8-sig"
)

print("\n4. INFORME GENERADO")
print(ruta_diagnostico)

print("\nAUDITORÍA TF-IDF FINALIZADA")
print("=" * 85)
```

## 9. Entrenamiento y comparación de modelos de clasificación

### 9.1. Objetivo

Desarrollar y comparar diferentes algoritmos de Machine Learning para clasificar reseñas de videojuegos según su recomendación.

### 9.2. Problema predictivo

Se plantea un problema de clasificación binaria:

- Clase 1: recomendación positiva.
- Clase 0: recomendación negativa.

La variable predictora principal será el contenido textual de la reseña, representado mediante TF-IDF.

### 9.3. Modelos seleccionados

Se evaluarán:

- Dummy Classifier como línea base.
- Regresión Logística.
- Regresión Logística con ponderación de clases.
- Linear Support Vector Machine.
- Linear Support Vector Machine con ponderación de clases.

### 9.4. Metodología de evaluación

Se utilizará validación cruzada estratificada agrupada en cinco particiones.

Los grupos definidos previamente impedirán que un mismo usuario o texto normalizado aparezca simultáneamente en entrenamiento y validación.

La vectorización TF-IDF se ajustará dentro de cada partición para evitar fuga de información.

### 9.5. Métricas

Se evaluarán:

- Accuracy.
- Balanced Accuracy.
- Precision Macro.
- Recall Macro.
- F1 Macro.
- Precision de la clase negativa.
- Recall de la clase negativa.
- F1 de la clase negativa.

Se prestará especial atención a la identificación de recomendaciones negativas debido al desequilibrio de clases.

### 9.6. Criterios de selección

Se compararán los resultados de validación cruzada, considerando tanto el desempeño general como la capacidad de identificar la clase minoritaria.

No se seleccionará un modelo únicamente por su exactitud.

### 9.7. Resultado esperado

Una comparación reproducible de modelos que permita seleccionar candidatos para el ajuste posterior y la evaluación final sobre el conjunto de prueba.


```python

# ============================================================
# PRIMERA COMPARACIÓN DE MODELOS ML
# ============================================================

print("=" * 90)
print("MACHINE LEARNING - VALIDACIÓN CRUZADA AGRUPADA")
print("=" * 90)

# ------------------------------------------------------------
# 1. PREPARAR DATOS
# ------------------------------------------------------------

X_cv = X_train_text.reset_index(drop=True)

y_cv = y_train.reset_index(drop=True).astype(int)

groups_cv = (
    train_data["grupo_particion"]
    .reset_index(drop=True)
    .to_numpy()
)

if not (
    len(X_cv) == len(y_cv) == len(groups_cv)
):
    raise ValueError(
        "Los textos, las etiquetas y los grupos no coinciden."
    )

# ------------------------------------------------------------
# 2. CONFIGURAR VALIDACIÓN CRUZADA
# ------------------------------------------------------------

cv_modelos = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# ------------------------------------------------------------
# 3. DEFINIR MÉTRICAS
# ------------------------------------------------------------

SCORING = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
    "f1_macro": "f1_macro",

    "precision_negativa": make_scorer(
        precision_score,
        pos_label=0,
        zero_division=0
    ),

    "recall_negativa": make_scorer(
        recall_score,
        pos_label=0,
        zero_division=0
    ),

    "f1_negativa": make_scorer(
        f1_score,
        pos_label=0,
        zero_division=0
    )
}

# ------------------------------------------------------------
# 4. CONFIGURAR TF-IDF DENTRO DE LOS PIPELINES
# ------------------------------------------------------------

def crear_vectorizador():

    return TfidfVectorizer(
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95,
        max_features=30000,
        sublinear_tf=True,
        norm="l2",
        strip_accents="unicode",
        lowercase=False,
        stop_words=None,
        dtype=np.float32,
        token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z']+\b"
    )


# ------------------------------------------------------------
# 5. DEFINIR MODELOS
# ------------------------------------------------------------

MODELOS = {
    "Dummy - mayoría": DummyClassifier(
        strategy="most_frequent"
    ),

    "Regresión Logística": Pipeline([
        ("tfidf", crear_vectorizador()),
        (
            "modelo",
            LogisticRegression(
                C=1.0,
                max_iter=2000,
                solver="liblinear",
                class_weight=None,
                random_state=42
            )
        )
    ]),

    "Regresión Logística - balanceada": Pipeline([
        ("tfidf", crear_vectorizador()),
        (
            "modelo",
            LogisticRegression(
                C=1.0,
                max_iter=2000,
                solver="liblinear",
                class_weight="balanced",
                random_state=42
            )
        )
    ]),

    "Linear SVM": Pipeline([
        ("tfidf", crear_vectorizador()),
        (
            "modelo",
            LinearSVC(
                C=1.0,
                class_weight=None,
                random_state=42
            )
        )
    ]),

    "Linear SVM - balanceada": Pipeline([
        ("tfidf", crear_vectorizador()),
        (
            "modelo",
            LinearSVC(
                C=1.0,
                class_weight="balanced",
                random_state=42
            )
        )
    ])
}

# Dummy recibe una matriz constante porque no utiliza texto.
X_dummy = np.zeros(
    (len(X_cv), 1),
    dtype=np.float32
)

# ------------------------------------------------------------
# 6. ENTRENAR Y EVALUAR
# ------------------------------------------------------------

resultados_cv = []
detalle_folds = []

for nombre, modelo in MODELOS.items():

    print("\n" + "-" * 75)
    print(f"MODELO: {nombre}")
    print("-" * 75)

    X_actual = (
        X_dummy
        if nombre == "Dummy - mayoría"
        else X_cv
    )

    resultado = cross_validate(
        estimator=modelo,
        X=X_actual,
        y=y_cv,
        groups=groups_cv,
        cv=cv_modelos,
        scoring=SCORING,
        return_train_score=False,
        return_estimator=False,
        n_jobs=1,
        error_score="raise"
    )

    fila = {
        "modelo": nombre,
        "tiempo_entrenamiento_promedio": np.mean(
            resultado["fit_time"]
        )
    }

    for metrica in SCORING:

        valores = resultado[f"test_{metrica}"]

        fila[f"{metrica}_media"] = valores.mean()
        fila[f"{metrica}_std"] = valores.std(ddof=1)

        for fold, valor in enumerate(valores, start=1):

            detalle_folds.append({
                "modelo": nombre,
                "fold": fold,
                "metrica": metrica,
                "valor": valor
            })

    resultados_cv.append(fila)

    print(
        f"F1 Macro promedio: "
        f"{fila['f1_macro_media']:.4f}"
    )

    print(
        f"Recall negativo promedio: "
        f"{fila['recall_negativa_media']:.4f}"
    )

    print(
        f"Balanced Accuracy promedio: "
        f"{fila['balanced_accuracy_media']:.4f}"
    )

# ------------------------------------------------------------
# 7. CONSOLIDAR RESULTADOS
# ------------------------------------------------------------

comparacion_modelos = pd.DataFrame(
    resultados_cv
).sort_values(
    ["f1_macro_media", "recall_negativa_media"],
    ascending=False
).reset_index(drop=True)

detalle_validacion = pd.DataFrame(
    detalle_folds
)

print("\n" + "=" * 90)
print("COMPARACIÓN GENERAL DE MODELOS")
print("=" * 90)

display(
    comparacion_modelos[
        [
            "modelo",
            "accuracy_media",
            "balanced_accuracy_media",
            "precision_macro_media",
            "recall_macro_media",
            "f1_macro_media",
            "precision_negativa_media",
            "recall_negativa_media",
            "f1_negativa_media",
            "tiempo_entrenamiento_promedio"
        ]
    ].round(4)
)

# ------------------------------------------------------------
# 8. EXPORTAR RESULTADOS
# ------------------------------------------------------------

comparacion_modelos.to_csv(
    REPORTS / "comparacion_modelos_validacion_cruzada.csv",
    index=False,
    encoding="utf-8-sig"
)

detalle_validacion.to_csv(
    REPORTS / "detalle_folds_validacion_cruzada.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nARCHIVOS GENERADOS")
print(REPORTS / "comparacion_modelos_validacion_cruzada.csv")
print(REPORTS / "detalle_folds_validacion_cruzada.csv")

print("\nVALIDACIÓN CRUZADA FINALIZADA")
print("=" * 90)
```

```python
# ============================================================
# ESTABILIDAD DE LOS MODELOS ENTRE FOLDS - ACA EVALUAMOS
# ============================================================

metricas_estabilidad = [
    "f1_macro",
    "balanced_accuracy",
    "recall_negativa",
    "f1_negativa"
]

columnas = ["modelo"]

for metrica in metricas_estabilidad:
    columnas.extend([
        f"{metrica}_media",
        f"{metrica}_std"
    ])

display(
    comparacion_modelos[columnas].round(4)
)
```


## 10. Optimización de hiperparámetros

### 10.1. Objetivo

Optimizar los dos modelos con mejor desempeño en la validación cruzada inicial:

- Regresión Logística balanceada.
- Linear Support Vector Machine balanceada.

### 10.2. Estrategia

Se utilizará GridSearchCV con validación cruzada estratificada agrupada de cinco folds.

Se evaluarán diferentes valores del parámetro C:

0.01, 0.1, 1, 10 y 100.

### 10.3. Criterio de selección

Se utilizará Balanced Accuracy como métrica principal de selección, debido al desequilibrio entre recomendaciones positivas y negativas.

También se registrarán F1 Macro, Recall de la clase negativa y F1 de la clase negativa.

### 10.4. Control metodológico

El conjunto de prueba permanecerá completamente reservado.

La búsqueda de hiperparámetros utilizará exclusivamente los datos de entrenamiento y sus grupos previamente definidos.


```python

# ============================================================
# STEAM ANALYTICS
# NOTEBOOK 2 - OPTIMIZACIÓN DE HIPERPARÁMETROS
# ============================================================

print("=" * 90)
print("OPTIMIZACIÓN DE HIPERPARÁMETROS")
print("=" * 90)

# ------------------------------------------------------------
# 1. MÉTRICAS DE EVALUACIÓN
# ------------------------------------------------------------

SCORING_OPT = {
    "balanced_accuracy": "balanced_accuracy",

    "f1_macro": make_scorer(
        f1_score,
        average="macro",
        zero_division=0
    ),

    "recall_negativa": make_scorer(
        recall_score,
        pos_label=0,
        zero_division=0
    ),

    "f1_negativa": make_scorer(
        f1_score,
        pos_label=0,
        zero_division=0
    )
}

# ------------------------------------------------------------
# 2. DEFINIR MODELOS Y GRILLAS
# ------------------------------------------------------------

modelos_optimizacion = {

    "Regresión Logística balanceada": {
        "pipeline": Pipeline([
            ("tfidf", crear_vectorizador()),
            (
                "modelo",
                LogisticRegression(
                    class_weight="balanced",
                    max_iter=2000,
                    solver="liblinear",
                    random_state=42
                )
            )
        ]),
        "parametros": {
            "modelo__C": [0.01, 0.1, 1, 10, 100]
        }
    },

    "Linear SVM balanceada": {
        "pipeline": Pipeline([
            ("tfidf", crear_vectorizador()),
            (
                "modelo",
                LinearSVC(
                    class_weight="balanced",
                    random_state=42
                )
            )
        ]),
        "parametros": {
            "modelo__C": [0.01, 0.1, 1, 10, 100]
        }
    }
}

# ------------------------------------------------------------
# 3. EJECUTAR BÚSQUEDA
# ------------------------------------------------------------

busquedas = {}
resultados_optimizacion = []

for nombre, configuracion in modelos_optimizacion.items():

    print("\n" + "-" * 75)
    print(f"OPTIMIZANDO: {nombre}")
    print("-" * 75)

    busqueda = GridSearchCV(
        estimator=configuracion["pipeline"],
        param_grid=configuracion["parametros"],
        scoring=SCORING_OPT,
        refit="balanced_accuracy",
        cv=cv_modelos,
        n_jobs=1,
        return_train_score=False,
        error_score="raise"
    )

    busqueda.fit(
        X_cv,
        y_cv,
        groups=groups_cv
    )

    busquedas[nombre] = busqueda

    print("Mejores parámetros:")
    print(busqueda.best_params_)

    print(
        "Mejor Balanced Accuracy:",
        round(busqueda.best_score_, 4)
    )

    # Registrar todos los resultados
    resultados = pd.DataFrame(
        busqueda.cv_results_
    )

    for _, fila in resultados.iterrows():

        resultados_optimizacion.append({
            "modelo": nombre,
            "C": fila["param_modelo__C"],
            "balanced_accuracy_media": fila[
                "mean_test_balanced_accuracy"
            ],
            "balanced_accuracy_std": fila[
                "std_test_balanced_accuracy"
            ],
            "f1_macro_media": fila[
                "mean_test_f1_macro"
            ],
            "recall_negativa_media": fila[
                "mean_test_recall_negativa"
            ],
            "f1_negativa_media": fila[
                "mean_test_f1_negativa"
            ]
        })

# ------------------------------------------------------------
# 4. CONSOLIDAR RESULTADOS
# ------------------------------------------------------------

resultados_optimizacion_df = pd.DataFrame(
    resultados_optimizacion
).sort_values(
    ["balanced_accuracy_media", "f1_macro_media"],
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 90)
print("RESULTADOS DE OPTIMIZACIÓN")
print("=" * 90)

display(
    resultados_optimizacion_df.round(4)
)

# ------------------------------------------------------------
# 5. IDENTIFICAR MODELOS OPTIMIZADOS
# ------------------------------------------------------------

mejores_modelos = {}

for nombre, busqueda in busquedas.items():

    mejores_modelos[nombre] = busqueda.best_estimator_

    print("\n", nombre)
    print("Parámetros:", busqueda.best_params_)
    print("Balanced Accuracy:", round(busqueda.best_score_, 4))

# ------------------------------------------------------------
# 6. EXPORTAR RESULTADOS
# ------------------------------------------------------------

resultados_optimizacion_df.to_csv(
    REPORTS / "optimizacion_hiperparametros.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nArchivo generado:")
print(REPORTS / "optimizacion_hiperparametros.csv")

print("\nOPTIMIZACIÓN FINALIZADA")
print("=" * 90)
```

```python

# ============================================================
# VERIFICACIÓN DE CONVERGENCIA
# ============================================================

print("=" * 90)
print("VERIFICACIÓN DE CONVERGENCIA Y ESTABILIDAD")
print("=" * 90)

# ------------------------------------------------------------
# 1. DEFINIR CANDIDATOS CON PARÁMETROS SELECCIONADOS
# ------------------------------------------------------------

candidatos_finales = {

    "Linear SVM balanceada C=0.1": Pipeline([
        ("tfidf", crear_vectorizador()),
        (
            "modelo",
            LinearSVC(
                C=0.1,
                class_weight="balanced",
                max_iter=20000,
                tol=1e-4,
                random_state=42
            )
        )
    ]),

    "Regresión Logística balanceada C=1": Pipeline([
        ("tfidf", crear_vectorizador()),
        (
            "modelo",
            LogisticRegression(
                C=1,
                class_weight="balanced",
                max_iter=2000,
                solver="liblinear",
                random_state=42
            )
        )
    ])
}

# ------------------------------------------------------------
# 2. VALIDACIÓN CRUZADA
# ------------------------------------------------------------

resultados_verificacion = []
detalles_verificacion = []

for nombre, modelo in candidatos_finales.items():

    print("\n" + "-" * 75)
    print(nombre)
    print("-" * 75)

    resultado = cross_validate(
        estimator=modelo,
        X=X_cv,
        y=y_cv,
        groups=groups_cv,
        cv=cv_modelos,
        scoring=SCORING_OPT,
        return_train_score=False,
        n_jobs=1,
        error_score="raise"
    )

    fila = {"modelo": nombre}

    for metrica in SCORING_OPT:

        valores = resultado[f"test_{metrica}"]

        fila[f"{metrica}_media"] = valores.mean()
        fila[f"{metrica}_std"] = valores.std(ddof=1)

        for fold, valor in enumerate(valores, start=1):

            detalles_verificacion.append({
                "modelo": nombre,
                "fold": fold,
                "metrica": metrica,
                "valor": valor
            })

    fila["tiempo_promedio"] = np.mean(
        resultado["fit_time"]
    )

    resultados_verificacion.append(fila)

    print(
        "Balanced Accuracy:",
        round(fila["balanced_accuracy_media"], 4)
    )

    print(
        "F1 Macro:",
        round(fila["f1_macro_media"], 4)
    )

    print(
        "Recall negativo:",
        round(fila["recall_negativa_media"], 4)
    )

# ------------------------------------------------------------
# 3. CONSOLIDAR RESULTADOS
# ------------------------------------------------------------

verificacion_modelos = pd.DataFrame(
    resultados_verificacion
).sort_values(
    "balanced_accuracy_media",
    ascending=False
).reset_index(drop=True)

detalle_verificacion = pd.DataFrame(
    detalles_verificacion
)

print("\n" + "=" * 90)
print("RESULTADOS DE VERIFICACIÓN")
print("=" * 90)

display(verificacion_modelos.round(4))

# ------------------------------------------------------------
# 4. EXPORTAR
# ------------------------------------------------------------

verificacion_modelos.to_csv(
    REPORTS / "verificacion_convergencia_modelos.csv",
    index=False,
    encoding="utf-8-sig"
)

detalle_verificacion.to_csv(
    REPORTS / "detalle_verificacion_convergencia.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nArchivos guardados:")
print(REPORTS / "verificacion_convergencia_modelos.csv")
print(REPORTS / "detalle_verificacion_convergencia.csv")

print("\nVERIFICACIÓN FINALIZADA")
print("=" * 90)
```


## 11. Evaluación final de modelos

### 11.1. Objetivo

Evaluar el rendimiento de los dos modelos seleccionados sobre el conjunto de prueba reservado.

### 11.2. Modelos evaluados

- Linear SVM balanceada, C=0.1.
- Regresión Logística balanceada, C=1.

### 11.3. Procedimiento

Cada modelo se entrenará utilizando la totalidad del conjunto de entrenamiento.

Posteriormente, se realizarán predicciones sobre el conjunto de prueba, sin modificar sus datos ni sus etiquetas.

### 11.4. Métricas

Se calcularán:

- Accuracy.
- Balanced Accuracy.
- Precision, Recall y F1 Macro.
- Precision, Recall y F1 para cada clase.
- Matriz de confusión.

### 11.5. Cobertura textual

Se realizará una evaluación complementaria sobre los documentos que tienen al menos una característica TF-IDF.

Los documentos sin características se conservarán en la evaluación general y se analizarán por separado.

### 11.6. Resultado esperado

Determinar el rendimiento de generalización de ambos modelos y documentar sus fortalezas y limitaciones.


```python

# ============================================================
# STEAM ANALYTICS
# NOTEBOOK 2 - EVALUACIÓN FINAL SOBRE TEST
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

print("=" * 90)
print("EVALUACIÓN FINAL - CONJUNTO DE PRUEBA")
print("=" * 90)

# ------------------------------------------------------------
# 1. PREPARAR CONJUNTOS
# ------------------------------------------------------------

X_train_final = X_train_text.reset_index(drop=True)
X_test_final = X_test_text.reset_index(drop=True)

y_train_final = y_train.reset_index(drop=True).astype(int)
y_test_final = y_test.reset_index(drop=True).astype(int)

test_final = test_data.reset_index(drop=True).copy()

if not (
    len(X_test_final) ==
    len(y_test_final) ==
    len(test_final)
):
    raise ValueError(
        "Los textos, etiquetas y registros de prueba no coinciden."
    )

# ------------------------------------------------------------
# 2. DEFINIR MODELOS
# ------------------------------------------------------------

modelos_finales = {

    "Linear SVM balanceada C=0.1": Pipeline([
        ("tfidf", crear_vectorizador()),
        (
            "modelo",
            LinearSVC(
                C=0.1,
                class_weight="balanced",
                max_iter=20000,
                tol=1e-4,
                random_state=42
            )
        )
    ]),

    "Regresión Logística balanceada C=1": Pipeline([
        ("tfidf", crear_vectorizador()),
        (
            "modelo",
            LogisticRegression(
                C=1,
                class_weight="balanced",
                max_iter=2000,
                solver="liblinear",
                random_state=42
            )
        )
    ])
}

# ------------------------------------------------------------
# 3. ENTRENAMIENTO Y PREDICCIONES
# ------------------------------------------------------------

resultados_test = []
predicciones_test = {}
matrices_confusion = {}

for nombre, modelo in modelos_finales.items():

    print("\n" + "-" * 75)
    print("MODELO:", nombre)
    print("-" * 75)

    # Entrenar con todo el conjunto de entrenamiento
    modelo.fit(
        X_train_final,
        y_train_final
    )

    # Predecir sobre test
    predicciones = modelo.predict(X_test_final)

    predicciones_test[nombre] = predicciones

    # Matriz de confusión
    matriz = confusion_matrix(
        y_test_final,
        predicciones,
        labels=[0, 1]
    )

    matrices_confusion[nombre] = matriz

    # Métricas generales
    metricas = {
        "modelo": nombre,
        "accuracy": accuracy_score(
            y_test_final, predicciones
        ),
        "balanced_accuracy": balanced_accuracy_score(
            y_test_final, predicciones
        ),
        "precision_macro": precision_score(
            y_test_final, predicciones,
            average="macro", zero_division=0
        ),
        "recall_macro": recall_score(
            y_test_final, predicciones,
            average="macro", zero_division=0
        ),
        "f1_macro": f1_score(
            y_test_final, predicciones,
            average="macro", zero_division=0
        ),
        "precision_negativa": precision_score(
            y_test_final, predicciones,
            pos_label=0, zero_division=0
        ),
        "recall_negativa": recall_score(
            y_test_final, predicciones,
            pos_label=0, zero_division=0
        ),
        "f1_negativa": f1_score(
            y_test_final, predicciones,
            pos_label=0, zero_division=0
        ),
        "precision_positiva": precision_score(
            y_test_final, predicciones,
            pos_label=1, zero_division=0
        ),
        "recall_positiva": recall_score(
            y_test_final, predicciones,
            pos_label=1, zero_division=0
        ),
        "f1_positiva": f1_score(
            y_test_final, predicciones,
            pos_label=1, zero_division=0
        )
    }

    resultados_test.append(metricas)

    print("\nMATRIZ DE CONFUSIÓN")
    print("Filas = clase real | Columnas = clase predicha")
    print(matriz)

    print("\nINFORME DE CLASIFICACIÓN")

    print(
        classification_report(
            y_test_final,
            predicciones,
            labels=[0, 1],
            target_names=[
                "Negativa",
                "Positiva"
            ],
            zero_division=0,
            digits=4
        )
    )

# ------------------------------------------------------------
# 4. COBERTURA TF-IDF EN TEST
# ------------------------------------------------------------

# Usamos el vectorizador de uno de los modelos.
# Ambos utilizan exactamente la misma configuración TF-IDF.

vectorizador_final = modelos_finales[
    "Linear SVM balanceada C=0.1"
].named_steps["tfidf"]

X_test_final_tfidf = vectorizador_final.transform(
    X_test_final
)

mascara_con_caracteristicas = (
    np.asarray(X_test_final_tfidf.getnnz(axis=1)).ravel() > 0
)

mascara_sin_caracteristicas = (
    ~mascara_con_caracteristicas
)

print("\n" + "=" * 90)
print("COBERTURA TF-IDF DEL CONJUNTO DE PRUEBA")
print("=" * 90)

print("Total:", len(X_test_final))
print(
    "Con características:",
    mascara_con_caracteristicas.sum()
)
print(
    "Sin características:",
    mascara_sin_caracteristicas.sum()
)
print(
    "Cobertura porcentual:",
    round(
        mascara_con_caracteristicas.mean() * 100,
        2
    ),
    "%"
)

# ------------------------------------------------------------
# 5. EVALUAR DOCUMENTOS CON CARACTERÍSTICAS
# ------------------------------------------------------------

resultados_cobertura = []

for nombre, predicciones in predicciones_test.items():

    metricas_cobertura = {
        "modelo": nombre,
        "documentos": int(
            mascara_con_caracteristicas.sum()
        ),
        "accuracy": accuracy_score(
            y_test_final[mascara_con_caracteristicas],
            predicciones[mascara_con_caracteristicas]
        ),
        "balanced_accuracy": balanced_accuracy_score(
            y_test_final[mascara_con_caracteristicas],
            predicciones[mascara_con_caracteristicas]
        ),
        "f1_macro": f1_score(
            y_test_final[mascara_con_caracteristicas],
            predicciones[mascara_con_caracteristicas],
            average="macro",
            zero_division=0
        ),
        "recall_negativa": recall_score(
            y_test_final[mascara_con_caracteristicas],
            predicciones[mascara_con_caracteristicas],
            pos_label=0,
            zero_division=0
        ),
        "f1_negativa": f1_score(
            y_test_final[mascara_con_caracteristicas],
            predicciones[mascara_con_caracteristicas],
            pos_label=0,
            zero_division=0
        )
    }

    resultados_cobertura.append(
        metricas_cobertura
    )

# ------------------------------------------------------------
# 6. CONSOLIDAR RESULTADOS
# ------------------------------------------------------------

comparacion_test = pd.DataFrame(
    resultados_test
)

comparacion_cobertura = pd.DataFrame(
    resultados_cobertura
)

print("\n" + "=" * 90)
print("RESULTADOS GENERALES - TEST COMPLETO")
print("=" * 90)

display(
    comparacion_test.round(4)
)

print("\n" + "=" * 90)
print("RESULTADOS - DOCUMENTOS CON CARACTERÍSTICAS TF-IDF")
print("=" * 90)

display(
    comparacion_cobertura.round(4)
)

# ------------------------------------------------------------
# 7. EXPORTAR MATRICES DE CONFUSIÓN
# ------------------------------------------------------------

for nombre, matriz in matrices_confusion.items():

    nombre_archivo = (
        nombre.lower()
        .replace(" ", "_")
        .replace("=", "")
    )

    pd.DataFrame(
        matriz,
        index=["Real_negativa", "Real_positiva"],
        columns=["Predicha_negativa", "Predicha_positiva"]
    ).to_csv(
        REPORTS / f"matriz_confusion_{nombre_archivo}.csv",
        encoding="utf-8-sig"
    )

# ------------------------------------------------------------
# 8. EXPORTAR MÉTRICAS
# ------------------------------------------------------------

comparacion_test.to_csv(
    REPORTS / "evaluacion_final_test.csv",
    index=False,
    encoding="utf-8-sig"
)

comparacion_cobertura.to_csv(
    REPORTS / "evaluacion_test_con_caracteristicas.csv",
    index=False,
    encoding="utf-8-sig"
)

# ------------------------------------------------------------
# 9. EXPORTAR PREDICCIONES
# ------------------------------------------------------------

columnas_identificacion = [
    columna
    for columna in [
        "recommendationid",
        "user_key",
        "game_key",
        "account_created_year",
        "voted_up"
    ]
    if columna in test_final.columns
]

predicciones_export = test_final[
    columnas_identificacion
].copy()

predicciones_export["tiene_caracteristicas_tfidf"] = (
    mascara_con_caracteristicas
)

for nombre, predicciones in predicciones_test.items():

    sufijo = (
        nombre.lower()
        .replace(" ", "_")
        .replace("=", "")
    )

    predicciones_export[
        f"prediccion_{sufijo}"
    ] = predicciones

predicciones_export.to_csv(
    REPORTS / "predicciones_modelos_test.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 90)
print("ARCHIVOS GENERADOS")
print("=" * 90)

for archivo in [
    "evaluacion_final_test.csv",
    "evaluacion_test_con_caracteristicas.csv",
    "predicciones_modelos_test.csv"
]:
    print(REPORTS / archivo)

print("\nEVALUACIÓN FINAL COMPLETADA")
print("=" * 90)
```

```python

# ============================================================
# CONTROL FINAL: BASELINE MAYORITARIO Y COBERTURA TF-IDF
# ============================================================

print("=" * 90)
print("CONTROL DE BASELINE Y DOCUMENTOS SIN CARACTERÍSTICAS")
print("=" * 90)

# ------------------------------------------------------------
# 1. BASELINE: PREDECIR SIEMPRE LA CLASE MAYORITARIA
# ------------------------------------------------------------

predicciones_dummy_test = np.ones(
    len(y_test_final),
    dtype=int
)

baseline_test = {
    "modelo": "Dummy - siempre positiva",
    "accuracy": accuracy_score(
        y_test_final, predicciones_dummy_test
    ),
    "balanced_accuracy": balanced_accuracy_score(
        y_test_final, predicciones_dummy_test
    ),
    "f1_macro": f1_score(
        y_test_final,
        predicciones_dummy_test,
        average="macro",
        zero_division=0
    ),
    "recall_negativa": recall_score(
        y_test_final,
        predicciones_dummy_test,
        pos_label=0,
        zero_division=0
    ),
    "f1_negativa": f1_score(
        y_test_final,
        predicciones_dummy_test,
        pos_label=0,
        zero_division=0
    )
}

comparacion_baseline = pd.DataFrame(
    [baseline_test] + [
        {
            "modelo": fila["modelo"],
            "accuracy": fila["accuracy"],
            "balanced_accuracy": fila["balanced_accuracy"],
            "f1_macro": fila["f1_macro"],
            "recall_negativa": fila["recall_negativa"],
            "f1_negativa": fila["f1_negativa"]
        }
        for _, fila in comparacion_test.iterrows()
    ]
)

print("\nCOMPARACIÓN CONTRA BASELINE")
display(comparacion_baseline.round(4))

# ------------------------------------------------------------
# 2. AUDITAR PREDICCIONES SIN CARACTERÍSTICAS
# ------------------------------------------------------------

print("\nPREDICCIONES SOBRE DOCUMENTOS SIN TF-IDF")

auditoria_sin_caracteristicas = []

for nombre, predicciones in predicciones_test.items():

    pred_sin = predicciones[
        mascara_sin_caracteristicas
    ]

    reales_sin = y_test_final[
        mascara_sin_caracteristicas
    ]

    auditoria_sin_caracteristicas.append({
        "modelo": nombre,
        "documentos": len(pred_sin),
        "predichas_negativas": int(
            (pred_sin == 0).sum()
        ),
        "predichas_positivas": int(
            (pred_sin == 1).sum()
        ),
        "reales_negativas": int(
            (reales_sin == 0).sum()
        ),
        "reales_positivas": int(
            (reales_sin == 1).sum()
        ),
        "aciertos": int(
            (pred_sin == reales_sin).sum()
        ),
        "accuracy": accuracy_score(
            reales_sin, pred_sin
        )
    })

auditoria_sin_caracteristicas_df = pd.DataFrame(
    auditoria_sin_caracteristicas
)

display(
    auditoria_sin_caracteristicas_df.round(4)
)

# ------------------------------------------------------------
# 3. EXPORTAR
# ------------------------------------------------------------

comparacion_baseline.to_csv(
    REPORTS / "comparacion_baseline_test.csv",
    index=False,
    encoding="utf-8-sig"
)

auditoria_sin_caracteristicas_df.to_csv(
    REPORTS / "auditoria_predicciones_sin_tfidf_test.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nArchivos generados:")
print(REPORTS / "comparacion_baseline_test.csv")
print(REPORTS / "auditoria_predicciones_sin_tfidf_test.csv")

print("\nCONTROL FINALIZADO")
print("=" * 90)
```

### 12.2 Entrenamiento y exportación del modelo definitivo

Se reconstruye el pipeline seleccionado mediante validación cruzada: TF-IDF + Linear SVM balanceada (C=0.1).

El entrenamiento se realiza sobre la totalidad del conjunto de entrenamiento, manteniendo el conjunto de test independiente y sin utilizarlo para ajustar el modelo.

El pipeline completo se exporta para su posterior reutilización en aplicaciones de análisis y predicción.

```python

# ============================================================
# 14. ENTRENAMIENTO Y EXPORTACION DEL MODELO DEFINITIVO
# LINEAR SVM BALANCEADA C=0.1
# ============================================================

import joblib
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from pathlib import Path

print("=" * 80)
print("ENTRENAMIENTO DEL MODELO DEFINITIVO")
print("=" * 80)

# ------------------------------------------------------------
# 1. Reconstruir entrenamiento desde corpus_modelo
# ------------------------------------------------------------

X_train_final = corpus_modelo.loc[train_idx, "review_clean"].fillna("").astype(str)
y_train_final = corpus_modelo.loc[train_idx, "voted_up"].astype(int)

X_test_verificacion = corpus_modelo.loc[test_idx, "review_clean"].fillna("").astype(str)

print("\nDATOS DE ENTRENAMIENTO")
print("Registros:", len(X_train_final))
print("Positivas:", (y_train_final == 1).sum())
print("Negativas:", (y_train_final == 0).sum())

# ------------------------------------------------------------
# 2. Construir pipeline definitivo
# ------------------------------------------------------------

pipeline_svm_final = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(1, 2),
            min_df=2,
            max_df=0.95,
            max_features=30000,
            sublinear_tf=True,
            norm="l2",
            strip_accents="unicode",
            lowercase=False,
            stop_words=None,
            dtype=np.float32,
            token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z']+\b"
        )
    ),
    (
        "modelo",
        LinearSVC(
            C=0.1,
            class_weight="balanced",
            random_state=42,
            max_iter=20000,
            tol=1e-4
        )
    )
])

# ------------------------------------------------------------
# 3. Entrenar
# ------------------------------------------------------------

print("\nENTRENANDO PIPELINE...")

pipeline_svm_final.fit(
    X_train_final,
    y_train_final
)

print("Entrenamiento finalizado.")

# ------------------------------------------------------------
# 4. Verificar estructura
# ------------------------------------------------------------

print("\nESTRUCTURA DEL PIPELINE")
print(pipeline_svm_final)

print("\nVOCABULARIO TF-IDF:")
print(len(pipeline_svm_final.named_steps["tfidf"].vocabulary_))

# ------------------------------------------------------------
# 5. Exportar modelo completo
# ------------------------------------------------------------

MODELS = Path(
    "/content/drive/MyDrive/Infraestructura-de-ciencia-de-datos/Analytics/Models"
)

MODELS.mkdir(parents=True, exist_ok=True)

ruta_modelo = MODELS / "steam_linear_svm_tfidf_final.joblib"

joblib.dump(
    pipeline_svm_final,
    ruta_modelo,
    compress=3
)

print("\nMODELO EXPORTADO:")
print(ruta_modelo)

# ------------------------------------------------------------
# 6. Verificar que se pueda cargar
# ------------------------------------------------------------

modelo_verificado = joblib.load(ruta_modelo)

print("\nVERIFICACION DE CARGA")

assert isinstance(modelo_verificado, Pipeline)
assert len(modelo_verificado.steps) == 2
assert isinstance(modelo_verificado.named_steps["modelo"], LinearSVC)

print("Pipeline cargado correctamente.")

# ------------------------------------------------------------
# 7. Prueba técnica de predicción
# ------------------------------------------------------------

prediccion_prueba = modelo_verificado.predict(
    X_train_final.iloc[:5]
)

print("\nPREDICCIONES DE PRUEBA TÉCNICA:")
print(prediccion_prueba)

print("\nEXPORTACION FINALIZADA")
print("=" * 80)
```

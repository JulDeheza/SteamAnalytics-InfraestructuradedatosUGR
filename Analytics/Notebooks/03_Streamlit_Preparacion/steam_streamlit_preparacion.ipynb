### 3. Validación de relaciones y claves del modelo dimensional

En esta etapa se verifica la integridad estructural del modelo dimensional construido en la capa Gold, con el objetivo de garantizar que las tablas puedan relacionarse correctamente durante el desarrollo de la aplicación Streamlit.

Se realizan cuatro controles principales:

1. **Unicidad de claves:** comprobar que las claves primarias de las dimensiones y tablas de hechos no contengan valores duplicados ni nulos.
2. **Integridad referencial:** verificar que las claves foráneas de las tablas de hechos y agregaciones tengan correspondencia con las claves primarias de sus respectivas dimensiones.
3. **Consistencia de agregaciones:** comprobar que no existan registros duplicados para una misma combinación de juego, año y mes, ni cantidades negativas de reseñas o novedades.
4. **Identificación de inconsistencias:** detectar registros sin correspondencia para evaluar su origen y determinar cómo deben tratarse en los indicadores y visualizaciones.

Estos controles son exclusivamente diagnósticos: no modifican, eliminan ni transforman los datos originales.

**Resultado esperado:** confirmar que el modelo dimensional mantiene relaciones consistentes y que sus tablas pueden utilizarse como fuente de información para los indicadores, filtros y visualizaciones de Steam Analytics.

````python

# ============================================================
# STEAM ANALYTICS - NOTEBOOK 3
# PASO 3: VALIDACION DE RELACIONES Y CLAVES
# ============================================================

print("=" * 70)
print("VALIDACION DE RELACIONES ENTRE TABLAS GOLD")
print("=" * 70)

# ------------------------------------------------------------
# 1. CLAVES PRIMARIAS: UNICIDAD
# ------------------------------------------------------------

claves = {
    "dim_usuarios": ("user_key", dim_usuarios),
    "dim_juegos": ("game_key", dim_juegos),
    "dim_fecha": ("date_key", dim_fecha),
    "dim_idioma": ("language_key", dim_idioma),
    "dim_hardware": ("hardware_key", dim_hardware),
    "fact_resenas": ("recommendationid", fact_resenas),
    "fact_novedades": ("gid", fact_novedades)
}

print("\n1. UNICIDAD DE CLAVES")

for nombre, (columna, df) in claves.items():

    duplicados = int(df[columna].duplicated().sum())
    nulos = int(df[columna].isna().sum())

    print(
        f"{nombre}.{columna}: "
        f"duplicados={duplicados} | nulos={nulos} | "
        f"{'OK' if duplicados == 0 and nulos == 0 else 'REVISAR'}"
    )

# ------------------------------------------------------------
# 2. INTEGRIDAD REFERENCIAL
# ------------------------------------------------------------

def verificar_fk(fact, columna_fk, dimension, columna_pk):

    valores_fact = set(fact[columna_fk].dropna().unique())
    valores_dim = set(dimension[columna_pk].dropna().unique())

    no_encontrados = valores_fact - valores_dim

    return {
        "clave_foranea": columna_fk,
        "registros": len(fact),
        "valores_distintos": len(valores_fact),
        "no_encontrados": len(no_encontrados),
        "estado": "OK" if len(no_encontrados) == 0 else "REVISAR"
    }

relaciones = [
    ("fact_resenas", "user_key", dim_usuarios, "user_key"),
    ("fact_resenas", "game_key", dim_juegos, "game_key"),
    ("fact_resenas", "date_key", dim_fecha, "date_key"),
    ("fact_resenas", "language_key", dim_idioma, "language_key"),
    ("fact_resenas", "hardware_key", dim_hardware, "hardware_key"),
    ("fact_novedades", "game_key", dim_juegos, "game_key"),
    ("fact_novedades", "date_key", dim_fecha, "date_key"),
    ("agg_juego_mes", "game_key", dim_juegos, "game_key")
]

print("\n2. INTEGRIDAD REFERENCIAL")

resultados_fk = []

for nombre, fk, dimension, pk in relaciones:

    resultado = verificar_fk(
        gold[nombre], fk, dimension, pk
    )

    resultado["tabla"] = nombre
    resultados_fk.append(resultado)

    print(
        f"{nombre}.{fk}: "
        f"{resultado['no_encontrados']} valores sin correspondencia "
        f"| {resultado['estado']}"
    )

resultados_fk = pd.DataFrame(resultados_fk)

# ------------------------------------------------------------
# 3. CONSISTENCIA DE AGREGACIONES
# ------------------------------------------------------------

print("\n3. VALIDACION DE AGREGACIONES")

duplicados_juego_mes = agg_juego_mes.duplicated(
    subset=["game_key", "year", "month"]
).sum()

valores_negativos = (
    (agg_juego_mes["cantidad_resenas"] < 0).sum()
    +
    (agg_juego_mes["cantidad_novedades"] < 0).sum()
)

print(f"Duplicados juego-mes: {duplicados_juego_mes}")
print(f"Valores negativos en cantidades: {valores_negativos}")

# ------------------------------------------------------------
# 4. RESUMEN
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RESUMEN DE VALIDACION")
print("=" * 70)

display(resultados_fk)

print("Validación de claves y relaciones finalizada.")
````

### 4. Preparación de las bases analíticas para Streamlit

A partir de las tablas Gold previamente validadas, se construyen las bases de trabajo que utilizará la aplicación interactiva.

El proceso integra las reseñas con información de usuarios, juegos e idiomas, conservando las claves originales y verificando que las uniones no alteren la cantidad de registros.

Se preparan dos niveles de análisis:

- **Análisis general:** contempla el conjunto completo de reseñas disponibles.
- **Análisis de cohortes:** considera exclusivamente usuarios cuya cuenta fue creada entre 2020 y 2025, utilizando `account_created_year` como criterio de segmentación.

También se generan indicadores descriptivos de recomendaciones, usuarios, juegos y actividad mensual.

Las transformaciones se realizan en memoria, sin modificar los archivos originales de la capa Gold.

````python

# ============================================================
# STEAM ANALYTICS - NOTEBOOK 3
# PASO 4: PREPARACION DE BASES ANALITICAS
# ============================================================

print("=" * 70)
print("PREPARACION DE BASES ANALITICAS")
print("=" * 70)

# ------------------------------------------------------------
# 1. INTEGRACION DE RESEÑAS Y DIMENSIONES
# ------------------------------------------------------------

columnas_usuarios = [
    "user_key",
    "account_created_year"
]

columnas_juegos = [
    "game_key",
    "name"
]

columnas_idiomas = [
    "language_key",
    "language_code"
]

resenas_dashboard = fact_resenas.merge(
    dim_usuarios[columnas_usuarios],
    on="user_key",
    how="left",
    validate="many_to_one"
)

resenas_dashboard = resenas_dashboard.merge(
    dim_juegos[columnas_juegos],
    on="game_key",
    how="left",
    validate="many_to_one"
)

resenas_dashboard = resenas_dashboard.merge(
    dim_idioma[columnas_idiomas],
    on="language_key",
    how="left",
    validate="many_to_one"
)

# ------------------------------------------------------------
# 2. VALIDACION DE LA INTEGRACION
# ------------------------------------------------------------

print("\n1. VALIDACION DE INTEGRACION")

print(f"Reseñas originales: {len(fact_resenas):,}")
print(f"Reseñas integradas: {len(resenas_dashboard):,}")
print(
    "Registros conservados:",
    "OK" if len(resenas_dashboard) == len(fact_resenas) else "ERROR"
)

assert len(resenas_dashboard) == len(fact_resenas), (
    "La integración modificó la cantidad de reseñas."
)

# ------------------------------------------------------------
# 3. SEPARACION DE POBLACIONES
# ------------------------------------------------------------

resenas_con_cohorte = resenas_dashboard[
    resenas_dashboard["account_created_year"].notna()
].copy()

resenas_cohortes = resenas_dashboard[
    resenas_dashboard["account_created_year"]
    .between(2020, 2025)
    .fillna(False)
].copy()

resenas_fuera_cohorte = resenas_dashboard[
    ~resenas_dashboard.index.isin(resenas_cohortes.index)
].copy()

print("\n2. SEGMENTACION DE POBLACIONES")

print(f"Total de reseñas: {len(resenas_dashboard):,}")
print(f"Reseñas con año de creación: {len(resenas_con_cohorte):,}")
print(f"Reseñas de cohortes 2020-2025: {len(resenas_cohortes):,}")
print(f"Reseñas fuera de cohortes principales: {len(resenas_fuera_cohorte):,}")

# ------------------------------------------------------------
# 4. INDICADORES GENERALES
# ------------------------------------------------------------

total_resenas = len(resenas_dashboard)

positivas = int(resenas_dashboard["voted_up"].sum())
negativas = int((~resenas_dashboard["voted_up"]).sum())

kpis_generales = pd.DataFrame([
    {
        "indicador": "Total de reseñas",
        "valor": total_resenas
    },
    {
        "indicador": "Usuarios con reseñas",
        "valor": resenas_dashboard["user_key"].nunique()
    },
    {
        "indicador": "Juegos con reseñas",
        "valor": resenas_dashboard["game_key"].nunique()
    },
    {
        "indicador": "Reseñas positivas",
        "valor": positivas
    },
    {
        "indicador": "Reseñas negativas",
        "valor": negativas
    },
    {
        "indicador": "Porcentaje positivo",
        "valor": round(positivas / total_resenas * 100, 2)
    },
    {
        "indicador": "Porcentaje negativo",
        "valor": round(negativas / total_resenas * 100, 2)
    }
])

print("\n3. INDICADORES GENERALES")
display(kpis_generales)

# ------------------------------------------------------------
# 5. INDICADORES POR COHORTE
# ------------------------------------------------------------

resumen_cohortes = (
    resenas_cohortes
    .groupby("account_created_year")
    .agg(
        cantidad_resenas=("recommendationid", "count"),
        usuarios=("user_key", "nunique"),
        juegos=("game_key", "nunique"),
        positivas=("voted_up", "sum")
    )
    .reset_index()
)

resumen_cohortes["negativas"] = (
    resumen_cohortes["cantidad_resenas"]
    - resumen_cohortes["positivas"]
)

resumen_cohortes["porcentaje_positivo"] = (
    resumen_cohortes["positivas"]
    / resumen_cohortes["cantidad_resenas"]
    * 100
).round(2)

resumen_cohortes["porcentaje_negativo"] = (
    resumen_cohortes["negativas"]
    / resumen_cohortes["cantidad_resenas"]
    * 100
).round(2)

resumen_cohortes = resumen_cohortes.sort_values(
    "account_created_year"
).reset_index(drop=True)

print("\n4. INDICADORES POR COHORTE")
display(resumen_cohortes)

# ------------------------------------------------------------
# 6. ACTIVIDAD MENSUAL
# ------------------------------------------------------------

actividad_mensual = (
    agg_juego_mes
    .groupby(["year", "month", "periodo"], as_index=False)
    .agg(
        cantidad_resenas=("cantidad_resenas", "sum"),
        cantidad_novedades=("cantidad_novedades", "sum")
    )
    .sort_values(["year", "month"])
    .reset_index(drop=True)
)

print("\n5. ACTIVIDAD MENSUAL")
print(f"Períodos disponibles: {len(actividad_mensual):,}")
display(actividad_mensual.head(10))

# ------------------------------------------------------------
# 7. RANKING DE JUEGOS
# ------------------------------------------------------------

ranking_juegos = (
    resenas_dashboard
    .groupby(["game_key", "name"], as_index=False)
    .agg(
        cantidad_resenas=("recommendationid", "count"),
        positivas=("voted_up", "sum")
    )
)

ranking_juegos["porcentaje_positivo"] = (
    ranking_juegos["positivas"]
    / ranking_juegos["cantidad_resenas"]
    * 100
).round(2)

ranking_juegos = ranking_juegos.sort_values(
    "cantidad_resenas",
    ascending=False
).reset_index(drop=True)

print("\n6. TOP 10 JUEGOS POR CANTIDAD DE RESEÑAS")
display(ranking_juegos.head(10))

# ------------------------------------------------------------
# 8. VALIDACION FINAL
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("VALIDACION DE BASES ANALITICAS")
print("=" * 70)

print(f"Base integrada: {len(resenas_dashboard):,} registros")
print(f"Base cohortes: {len(resenas_cohortes):,} registros")
print(f"Indicadores generales: {len(kpis_generales)}")
print(f"Cohortes identificadas: {len(resumen_cohortes)}")
print(f"Períodos mensuales: {len(actividad_mensual)}")
print(f"Juegos en ranking: {len(ranking_juegos)}")

print("\nBases analíticas preparadas correctamente.")
````

### 5. Exportación de recursos analíticos para Streamlit

En esta etapa se exportan las bases analíticas construidas a partir de Gold, en formato Parquet, para utilizarlas posteriormente desde la aplicación Streamlit.

Se conservarán las siguientes estructuras:

- Base integrada de reseñas.
- Base de reseñas correspondiente a las cohortes 2020–2025.
- Indicadores generales.
- Resumen de indicadores por cohorte.
- Actividad mensual de juegos.
- Ranking de juegos según cantidad de reseñas.

Los archivos se almacenan en el directorio `Analytics/Streamlit/Data`, manteniendo separados los datos preparados, los modelos predictivos y los informes del proyecto.

Finalmente, se verifica que los archivos exportados puedan recuperarse correctamente y conserven sus dimensiones originales.

````python

# ============================================================
# STEAM ANALYTICS - NOTEBOOK 3
# PASO 5: EXPORTACION DE BASES ANALITICAS
# ============================================================

print("=" * 70)
print("EXPORTACION DE BASES ANALITICAS PARA STREAMLIT")
print("=" * 70)

# ------------------------------------------------------------
# 1. DIRECTORIO DE EXPORTACION
# ------------------------------------------------------------

STREAMLIT_DATA_DIR = STREAMLIT_DIR / "Data"
STREAMLIT_DATA_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. BASES A EXPORTAR
# ------------------------------------------------------------

bases_exportacion = {
    "resenas_dashboard": resenas_dashboard,
    "resenas_cohortes": resenas_cohortes,
    "kpis_generales": kpis_generales,
    "resumen_cohortes": resumen_cohortes,
    "actividad_mensual": actividad_mensual,
    "ranking_juegos": ranking_juegos
}

# ------------------------------------------------------------
# 3. EXPORTACION PARQUET
# ------------------------------------------------------------

print("\n1. EXPORTANDO ARCHIVOS")

for nombre, df in bases_exportacion.items():

    ruta = STREAMLIT_DATA_DIR / f"{nombre}.parquet"

    df.to_parquet(
        ruta,
        index=False
    )

    print(
        f"{nombre}.parquet | "
        f"{len(df):,} registros | "
        f"{len(df.columns)} columnas | "
        f"OK"
    )

# ------------------------------------------------------------
# 4. VERIFICACION DE ARCHIVOS
# ------------------------------------------------------------

print("\n2. VERIFICACION DE EXPORTACIONES")

resultados_exportacion = []

for nombre, df_original in bases_exportacion.items():

    ruta = STREAMLIT_DATA_DIR / f"{nombre}.parquet"

    df_recuperado = pd.read_parquet(ruta)

    dimensiones_ok = (
        df_original.shape == df_recuperado.shape
    )

    resultados_exportacion.append({
        "archivo": f"{nombre}.parquet",
        "filas_originales": len(df_original),
        "filas_recuperadas": len(df_recuperado),
        "columnas_originales": len(df_original.columns),
        "columnas_recuperadas": len(df_recuperado.columns),
        "verificacion": "OK" if dimensiones_ok else "ERROR"
    })

resultados_exportacion = pd.DataFrame(
    resultados_exportacion
)

display(resultados_exportacion)

# ------------------------------------------------------------
# 5. RESUMEN FINAL
# ------------------------------------------------------------

total_archivos = len(resultados_exportacion)

archivos_ok = (
    resultados_exportacion["verificacion"] == "OK"
).sum()

print("\n" + "=" * 70)
print("RESULTADO DE EXPORTACION")
print("=" * 70)

print(f"Archivos generados: {total_archivos}")
print(f"Archivos verificados: {archivos_ok}")
print(f"Directorio: {STREAMLIT_DATA_DIR}")

print(
    "\nEstado:",
    "OK" if total_archivos == archivos_ok else "REVISAR"
)

print("\nEXPORTACION FINALIZADA")
````

````python

# ============================================================
# PASO 7: ESTRUCTURA Y PAGINAS DE LA APLICACION STREAMLIT
# STEAM ANALYTICS
# ============================================================

import ast

# ============================================================
# 1. DIRECTORIOS PRINCIPALES
# ============================================================

APP_DIR = STREAMLIT_DIR
PAGES_DIR = APP_DIR / "pages"
UTILS_DIR = APP_DIR / "utils"
ASSETS_DIR = APP_DIR / "assets"

for directorio in [APP_DIR, PAGES_DIR, UTILS_DIR, ASSETS_DIR]:
    directorio.mkdir(parents=True, exist_ok=True)

archivos_aplicacion = [
    APP_DIR / "app.py",
    APP_DIR / "requirements.txt",
    APP_DIR / "README.md",
    PAGES_DIR / "01_Analisis_General.py",
    PAGES_DIR / "02_Analisis_Cohortes.py",
    PAGES_DIR / "03_Exploracion_Juegos.py",
    PAGES_DIR / "04_Analisis_Sentimiento.py",
    UTILS_DIR / "__init__.py",
    UTILS_DIR / "data_loader.py",
    UTILS_DIR / "visualizaciones.py",
    UTILS_DIR / "predicciones.py"
]

for archivo in archivos_aplicacion:
    archivo.touch(exist_ok=True)


# ============================================================
# 2. NAVEGACION COMPARTIDA
# ============================================================

NAVIGATION_CODE = '''
import streamlit as st


def mostrar_navegacion():

    with st.sidebar:

        st.title("🎮 Steam Analytics")

        st.caption(
            "Business Intelligence | Gaming Market Analytics"
        )

        st.divider()

        st.subheader("PLATAFORMA ANALITICA")

        st.page_link(
            "app.py",
            label="Inicio",
            icon="🏠"
        )

        st.page_link(
            "pages/01_Analisis_General.py",
            label="Análisis general",
            icon="📊"
        )

        st.page_link(
            "pages/02_Analisis_Cohortes.py",
            label="Cohortes y usuarios",
            icon="👥"
        )

        st.page_link(
            "pages/03_Exploracion_Juegos.py",
            label="Exploración de juegos",
            icon="🎮"
        )

        st.page_link(
            "pages/04_Analisis_Sentimiento.py",
            label="Análisis de sentimiento",
            icon="💬"
        )

        st.divider()

        st.subheader("INFORMACION DEL PROYECTO")

        st.caption(
            "Fuentes de datos, metodología y documentación "
            "del modelo disponibles en Inicio."
        )

        st.divider()

        st.caption(
            "Steam Analytics | Data Science & Machine Learning"
        )
'''

(UTILS_DIR / "__init__.py").write_text(
    NAVIGATION_CODE.strip() + "\n",
    encoding="utf-8"
)

print("Navegación compartida generada correctamente.")


# ============================================================
# PAGINA 1: ANALISIS GENERAL
# ============================================================

pagina_general = '''
import streamlit as st
import pandas as pd

from utils import mostrar_navegacion
from utils.data_loader import cargar_datasets
from utils.visualizaciones import (
    grafico_sentimiento,
    grafico_actividad_mensual,
    grafico_ranking_juegos
)

st.set_page_config(
    page_title="Análisis general",
    page_icon="📊",
    layout="wide"
)

mostrar_navegacion()

# ============================================================
# 1. CARGA Y PREPARACION
# ============================================================

datos = cargar_datasets()

resenas = datos["resenas_dashboard"].copy()

resenas["fecha_analisis"] = pd.to_datetime(
    resenas["timestamp_created_parsed"],
    errors="coerce",
    utc=True
)

resenas = resenas.dropna(
    subset=["fecha_analisis"]
)

st.title("📊 Análisis general")

st.markdown(
    """
    Diagnóstico general del mercado de videojuegos de Steam.

    Identificamos señales de demanda, aceptación, participación
    y oportunidades de mejora para desarrolladores, publishers
    e inversores.
    """
)

st.divider()

if resenas.empty:
    st.warning("No hay reseñas con fechas válidas.")
    st.stop()

# ============================================================
# 2. FILTROS
# ============================================================

st.subheader("Filtros de análisis")

fecha_min = resenas["fecha_analisis"].min().date()
fecha_max = resenas["fecha_analisis"].max().date()

col1, col2 = st.columns(2)

with col1:
    periodo = st.date_input(
        "Período de publicación de reseñas",
        value=(fecha_min, fecha_max),
        min_value=fecha_min,
        max_value=fecha_max,
        format="DD/MM/YYYY"
    )

with col2:
    idiomas = sorted(
        resenas["language_code"].dropna().unique().tolist()
    )

    idioma = st.selectbox(
        "Idioma",
        ["Todos"] + idiomas
    )

col3, col4 = st.columns(2)

with col3:
    juegos = sorted(
        resenas["name"].dropna().unique().tolist()
    )

    juego = st.selectbox(
        "Videojuego",
        ["Todos"] + juegos
    )

with col4:
    tipo = st.selectbox(
        "Tipo de recomendación",
        ["Todas", "Positivas", "Negativas"]
    )

# ============================================================
# 3. APLICAR FILTROS
# ============================================================

if isinstance(periodo, (tuple, list)) and len(periodo) == 2:

    fecha_inicio, fecha_fin = periodo

    if fecha_inicio > fecha_fin:
        st.error("El período seleccionado no es válido.")
        st.stop()

    filtradas = resenas[
        (resenas["fecha_analisis"].dt.date >= fecha_inicio) &
        (resenas["fecha_analisis"].dt.date <= fecha_fin)
    ].copy()

else:
    filtradas = resenas[
        resenas["fecha_analisis"].dt.date == periodo
    ].copy()

if idioma != "Todos":
    filtradas = filtradas[
        filtradas["language_code"] == idioma
    ]

if juego != "Todos":
    filtradas = filtradas[
        filtradas["name"] == juego
    ]

if tipo == "Positivas":
    filtradas = filtradas[
        filtradas["voted_up"] == True
    ]

elif tipo == "Negativas":
    filtradas = filtradas[
        filtradas["voted_up"] == False
    ]

if filtradas.empty:
    st.warning("No existen registros para los filtros seleccionados.")
    st.stop()

# ============================================================
# 4. INDICADORES
# ============================================================

st.divider()

st.subheader("Indicadores del período seleccionado")

total_resenas = len(filtradas)

usuarios = filtradas["steamid"].nunique()
cantidad_juegos = filtradas["name"].nunique()

valoraciones = filtradas["voted_up"].dropna()

positivas = int((valoraciones == True).sum())
negativas = int((valoraciones == False).sum())

total_valoraciones = positivas + negativas

porcentaje_positivo = (
    positivas / total_valoraciones * 100
    if total_valoraciones else 0
)

porcentaje_negativo = (
    negativas / total_valoraciones * 100
    if total_valoraciones else 0
)

col1, col2, col3 = st.columns(3)

col1.metric("Reseñas", f"{total_resenas:,}")
col2.metric("Usuarios", f"{usuarios:,}")
col3.metric("Juegos", f"{cantidad_juegos:,}")

col1, col2 = st.columns(2)

col1.metric(
    "Aceptación positiva",
    f"{porcentaje_positivo:.2f}%",
    f"{positivas:,} reseñas"
)

col2.metric(
    "Aceptación negativa",
    f"{porcentaje_negativo:.2f}%",
    f"{negativas:,} reseñas"
)

st.caption(
    "Las valoraciones se calculan sobre las reseñas con "
    "recomendación válida."
)

# ============================================================
# 5. DISTRIBUCION DE VALORACIONES
# ============================================================

st.divider()

st.subheader("Distribución de recomendaciones")

col1, col2 = st.columns(2)

with col1:
    st.plotly_chart(
        grafico_sentimiento(filtradas),
        use_container_width=True
    )

with col2:
    distribucion = pd.DataFrame({
        "Tipo": ["Positivas", "Negativas"],
        "Cantidad": [positivas, negativas]
    })

    st.dataframe(
        distribucion,
        use_container_width=True,
        hide_index=True
    )

# ============================================================
# 6. EVOLUCION MENSUAL
# ============================================================

st.divider()

st.subheader("Evolución mensual de reseñas")

actividad = (
    filtradas
    .assign(
        year=filtradas["fecha_analisis"].dt.year,
        month=filtradas["fecha_analisis"].dt.month,
        periodo=filtradas["fecha_analisis"].dt.strftime("%Y-%m")
    )
    .groupby(
        ["year", "month", "periodo"],
        as_index=False
    )
    .size()
    .rename(columns={"size": "cantidad_resenas"})
)

if not actividad.empty:

    st.plotly_chart(
        grafico_actividad_mensual(actividad),
        use_container_width=True
    )

    pico = actividad.loc[
        actividad["cantidad_resenas"].idxmax()
    ]

    st.caption(
        f"Mayor actividad: {pico['periodo']} "
        f"({pico['cantidad_resenas']:,} reseñas)."
    )

# ============================================================
# 7. RANKING DE JUEGOS
# ============================================================

st.divider()

st.subheader("Ranking de videojuegos")

ranking = (
    filtradas
    .groupby("name", as_index=False)
    .agg(
        cantidad_resenas=("recommendationid", "count"),
        positivas=("voted_up", lambda x: (x == True).sum()),
        negativas=("voted_up", lambda x: (x == False).sum()),
        total_valoraciones=("voted_up", "count")
    )
)

ranking["porcentaje_positivo"] = (
    ranking["positivas"]
    .div(ranking["total_valoraciones"].where(
        ranking["total_valoraciones"] > 0
    ))
    .mul(100)
)

ranking["porcentaje_negativo"] = (
    ranking["negativas"]
    .div(ranking["total_valoraciones"].where(
        ranking["total_valoraciones"] > 0
    ))
    .mul(100)
)

limite = st.slider(
    "Cantidad de juegos",
    min_value=5,
    max_value=30,
    value=10
)

ranking = ranking.sort_values(
    "cantidad_resenas",
    ascending=False
)

st.plotly_chart(
    grafico_ranking_juegos(ranking, limite),
    use_container_width=True
)

st.dataframe(
    ranking.head(limite),
    use_container_width=True,
    hide_index=True
)

# ============================================================
# 8. DIAGNOSTICO EMPRESARIAL
# ============================================================

st.divider()

st.header("Diagnóstico empresarial")

if porcentaje_negativo >= 30:

    st.error(
        f"""
        Riesgo elevado: el {porcentaje_negativo:.2f}% de las
        valoraciones son negativas.

        Acción: investigar problemas recurrentes y priorizar
        mejoras que puedan aumentar la satisfacción.
        """
    )

elif porcentaje_negativo >= 20:

    st.warning(
        f"""
        Señal de atención: el {porcentaje_negativo:.2f}% de las
        valoraciones son negativas.

        Acción: analizar las causas de insatisfacción y evaluar
        mejoras en rendimiento, contenido y experiencia.
        """
    )

elif porcentaje_positivo >= 80:

    st.success(
        f"""
        Señal favorable: el {porcentaje_positivo:.2f}% de las
        valoraciones son positivas.

        Acción: identificar los atributos mejor valorados y
        estudiar estrategias de comunidad y fidelización.
        """
    )

else:

    st.info(
        "Aceptación intermedia. Conviene analizar las causas "
        "de las valoraciones positivas y negativas."
    )

st.caption(
    "Los umbrales son criterios exploratorios, no estándares universales."
)

# ============================================================
# 9. OPORTUNIDADES POR JUEGO
# ============================================================

st.divider()

st.subheader("Oportunidades detectadas")

if not ranking.empty:

    col1, col2 = st.columns(2)

    with col1:

        mayor_volumen = ranking.iloc[0]

        st.markdown("**Mayor volumen de reseñas**")

        st.write(mayor_volumen["name"])

        st.metric(
            "Reseñas",
            f"{mayor_volumen['cantidad_resenas']:,}"
        )

        st.caption(
            "Investigar los factores que generan interés y "
            "participación para identificar oportunidades de posicionamiento."
        )

    with col2:

        muestra = ranking[
            ranking["total_valoraciones"] >= 30
        ].dropna(
            subset=["porcentaje_positivo"]
        )

        if not muestra.empty:

            mejor = muestra.loc[
                muestra["porcentaje_positivo"].idxmax()
            ]

            st.markdown("**Mayor aceptación con muestra suficiente**")

            st.write(mejor["name"])

            st.metric(
                "Positividad",
                f"{mejor['porcentaje_positivo']:.2f}%"
            )

            st.caption(
                "Resultado basado en al menos 30 valoraciones válidas."
            )

        else:

            st.info(
                "No hay juegos con una muestra mínima de 30 valoraciones."
            )

# ============================================================
# 10. RECOMENDACIONES POR PUBLICO
# ============================================================

st.divider()

st.subheader("Recomendaciones empresariales")

tab1, tab2, tab3 = st.tabs([
    "Desarrolladores",
    "Publishers",
    "Inversores"
])

with tab1:

    st.markdown(
        """
        - Priorizar problemas recurrentes en las reseñas negativas.
        - Identificar características asociadas con la aceptación.
        - Medir cambios de percepción después de actualizaciones.
        - Evaluar oportunidades de mejora de experiencia.
        """
    )

with tab2:

    st.markdown(
        """
        - Identificar productos con interés y buena aceptación.
        - Investigar riesgos reputacionales antes de ampliar campañas.
        - Comparar videojuegos y segmentos de mercado.
        - Contrastar participación con conversión y ventas.
        """
    )

with tab3:

    st.markdown(
        """
        - Utilizar las reseñas como señal complementaria de demanda.
        - Investigar tendencias de participación y aceptación.
        - Contrastar indicadores con ingresos, costos y ventas.
        - Evaluar la sostenibilidad comercial de cada producto.
        """
    )

st.warning(
    """
    Las reseñas no representan directamente ventas, ingresos,
    beneficios ni rentabilidad. Para estimarlos se necesitan
    datos económicos y comerciales adicionales.
    """
)

# ============================================================
# 11. RESUMEN
# ============================================================

st.divider()

st.subheader("Resumen del análisis")

col1, col2, col3 = st.columns(3)

col1.metric("Reseñas analizadas", f"{total_resenas:,}")
col2.metric("Aceptación positiva", f"{porcentaje_positivo:.2f}%")
col3.metric("Juegos analizados", f"{cantidad_juegos:,}")

st.caption(
    "Resultados dinámicos según los filtros seleccionados."
)
'''

(PAGES_DIR / "01_Analisis_General.py").write_text(
    pagina_general.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# PAGINA 2: ANALISIS DE COHORTES
# ============================================================

pagina_cohortes = '''
import streamlit as st
import pandas as pd
import plotly.express as px

from utils import mostrar_navegacion
from utils.data_loader import cargar_datasets
from utils.visualizaciones import grafico_cohortes

st.set_page_config(
    page_title="Análisis de cohortes",
    page_icon="👥",
    layout="wide"
)

mostrar_navegacion()

# ============================================================
# 1. CARGA Y VALIDACION
# ============================================================

datos = cargar_datasets()

cohortes = datos["resenas_cohortes"].copy()
resumen = datos["resumen_cohortes"].copy()

st.title("👥 Análisis de cohortes")

st.markdown(
    """
    Comparación del comportamiento de los usuarios según
    el año de creación de su cuenta de Steam.

    Buscamos identificar diferencias en participación,
    aceptación y oportunidades de segmentación comercial.
    """
)

ANIOS_COHORTE = list(range(2020, 2026))

columnas_requeridas = [
    "account_created_year",
    "cantidad_resenas",
    "usuarios",
    "juegos",
    "positivas",
    "negativas",
    "porcentaje_positivo",
    "porcentaje_negativo"
]

faltantes = [
    c for c in columnas_requeridas
    if c not in resumen.columns
]

if faltantes:
    st.error(
        "Faltan columnas del resumen: " + ", ".join(faltantes)
    )
    st.stop()

if "account_created_year" not in cohortes.columns:
    st.error("Falta account_created_year en las reseñas.")
    st.stop()

resumen = resumen[
    resumen["account_created_year"].isin(ANIOS_COHORTE)
].copy()

cohortes = cohortes[
    cohortes["account_created_year"].isin(ANIOS_COHORTE)
].copy()

if resumen.empty or cohortes.empty:
    st.warning("No hay datos disponibles para las cohortes 2020–2025.")
    st.stop()

for columna in columnas_requeridas[1:]:
    resumen[columna] = pd.to_numeric(
        resumen[columna],
        errors="coerce"
    ).fillna(0)

resumen["account_created_year"] = (
    resumen["account_created_year"].astype(int)
)

resumen = resumen.sort_values(
    "account_created_year"
).reset_index(drop=True)

# ============================================================
# 2. INDICADORES GENERALES
# ============================================================

st.info(
    "Análisis principal limitado a las seis cohortes 2020–2025."
)

total_resenas = int(resumen["cantidad_resenas"].sum())
total_usuarios = int(resumen["usuarios"].sum())
total_positivas = int(resumen["positivas"].sum())
total_negativas = int(resumen["negativas"].sum())

total_valoraciones = total_positivas + total_negativas

porcentaje_positivo = (
    total_positivas / total_valoraciones * 100
    if total_valoraciones else 0
)

porcentaje_negativo = (
    total_negativas / total_valoraciones * 100
    if total_valoraciones else 0
)

promedio_usuario = (
    total_resenas / total_usuarios
    if total_usuarios else 0
)

st.subheader("Indicadores generales")

col1, col2, col3 = st.columns(3)

col1.metric("Reseñas", f"{total_resenas:,}")
col2.metric("Usuarios", f"{total_usuarios:,}")
col3.metric("Juegos por cohorte", f"{int(resumen['juegos'].sum()):,}")

col1, col2, col3 = st.columns(3)

col1.metric("Positividad", f"{porcentaje_positivo:.2f}%")
col2.metric("Negatividad", f"{porcentaje_negativo:.2f}%")
col3.metric("Reseñas por usuario", f"{promedio_usuario:.2f}")

st.caption(
    "Los usuarios se contabilizan por cohorte. No se suman como "
    "usuarios únicos globales si un usuario puede aparecer "
    "en más de una cohorte."
)

# ============================================================
# 3. TABLA COMPARATIVA
# ============================================================

st.divider()

st.subheader("Indicadores por cohorte")

tabla = resumen.copy()

tabla["resenas_por_usuario"] = (
    tabla["cantidad_resenas"]
    .div(tabla["usuarios"].where(tabla["usuarios"] > 0))
)

tabla_presentacion = tabla.rename(
    columns={
        "account_created_year": "Cohorte",
        "cantidad_resenas": "Reseñas",
        "usuarios": "Usuarios",
        "juegos": "Juegos",
        "positivas": "Positivas",
        "negativas": "Negativas",
        "porcentaje_positivo": "% positivo",
        "porcentaje_negativo": "% negativo",
        "resenas_por_usuario": "Reseñas por usuario"
    }
)

st.dataframe(
    tabla_presentacion.round(2),
    use_container_width=True,
    hide_index=True
)

# ============================================================
# 4. PARTICIPACION POR COHORTE
# ============================================================

st.divider()

st.subheader("Participación de las cohortes")

fig_participacion = px.bar(
    resumen,
    x="account_created_year",
    y="cantidad_resenas",
    text="cantidad_resenas",
    labels={
        "account_created_year": "Año de creación de cuenta",
        "cantidad_resenas": "Reseñas"
    },
    title="Volumen de reseñas por cohorte"
)

fig_participacion.update_traces(
    textposition="outside"
)

fig_participacion.update_layout(
    xaxis=dict(tickmode="linear", dtick=1),
    showlegend=False
)

st.plotly_chart(
    fig_participacion,
    use_container_width=True
)

# ============================================================
# 5. ACTIVIDAD POR USUARIO
# ============================================================

st.divider()

st.subheader("Actividad promedio por usuario")

fig_actividad = px.bar(
    tabla,
    x="account_created_year",
    y="resenas_por_usuario",
    text="resenas_por_usuario",
    labels={
        "account_created_year": "Cohorte",
        "resenas_por_usuario": "Reseñas por usuario"
    },
    title="Promedio de reseñas por usuario"
)

fig_actividad.update_traces(
    texttemplate="%{text:.2f}",
    textposition="outside"
)

fig_actividad.update_layout(
    xaxis=dict(tickmode="linear", dtick=1),
    showlegend=False
)

st.plotly_chart(
    fig_actividad,
    use_container_width=True
)

# ============================================================
# 6. ACEPTACION POR COHORTE
# ============================================================

st.divider()

st.subheader("Evolución de la aceptación")

st.plotly_chart(
    grafico_cohortes(resumen),
    use_container_width=True
)

fig_valoraciones = px.line(
    resumen,
    x="account_created_year",
    y=["porcentaje_positivo", "porcentaje_negativo"],
    markers=True,
    labels={
        "account_created_year": "Cohorte",
        "value": "Porcentaje",
        "variable": "Valoración"
    },
    title="Porcentaje de valoraciones por cohorte"
)

fig_valoraciones.update_layout(
    xaxis=dict(tickmode="linear", dtick=1),
    yaxis_title="Porcentaje (%)"
)

st.plotly_chart(
    fig_valoraciones,
    use_container_width=True
)

primera = resumen.iloc[0]
ultima = resumen.iloc[-1]

diferencia = (
    ultima["porcentaje_positivo"]
    - primera["porcentaje_positivo"]
)

st.metric(
    "Diferencia de positividad entre primera y última cohorte",
    f"{diferencia:+.2f} puntos porcentuales"
)

st.caption(
    "La diferencia es descriptiva y no implica causalidad."
)

# ============================================================
# 7. COMPARACION ENTRE DOS COHORTES
# ============================================================

st.divider()

st.subheader("Comparación detallada")

anios = sorted(
    resumen["account_created_year"].unique().tolist()
)

col1, col2 = st.columns(2)

with col1:
    cohorte_a = st.selectbox(
        "Primera cohorte",
        anios,
        index=0
    )

with col2:
    cohorte_b = st.selectbox(
        "Segunda cohorte",
        anios,
        index=len(anios) - 1,
        key="cohorte_comparacion_b"
    )

fila_a = resumen[
    resumen["account_created_year"] == cohorte_a
].iloc[0]

fila_b = resumen[
    resumen["account_created_year"] == cohorte_b
].iloc[0]

comparacion = []

for nombre, columna in [
    ("Reseñas", "cantidad_resenas"),
    ("Usuarios", "usuarios"),
    ("Juegos", "juegos"),
    ("Positivas", "positivas"),
    ("Negativas", "negativas"),
    ("% positivo", "porcentaje_positivo"),
    ("% negativo", "porcentaje_negativo")
]:

    valor_a = fila_a[columna]
    valor_b = fila_b[columna]

    comparacion.append({
        "Indicador": nombre,
        str(cohorte_a): round(valor_a, 2),
        str(cohorte_b): round(valor_b, 2),
        "Diferencia": round(valor_b - valor_a, 2)
    })

st.dataframe(
    pd.DataFrame(comparacion),
    use_container_width=True,
    hide_index=True
)

# ============================================================
# 8. EXPLORACION DE UNA COHORTE
# ============================================================

st.divider()

st.subheader("Exploración de usuarios y reseñas")

anio = st.selectbox(
    "Seleccionar cohorte",
    anios,
    key="cohorte_exploracion"
)

filtradas = cohortes[
    cohortes["account_created_year"] == anio
].copy()

if filtradas.empty:

    st.warning("No hay reseñas para la cohorte seleccionada.")

else:

    positivas = int(
        filtradas["voted_up"].eq(True).sum()
    )

    negativas = int(
        filtradas["voted_up"].eq(False).sum()
    )

    col1, col2, col3 = st.columns(3)

    col1.metric("Reseñas", f"{len(filtradas):,}")
    col2.metric("Positivas", f"{positivas:,}")
    col3.metric("Negativas", f"{negativas:,}")

    distribucion = pd.DataFrame({
        "Tipo": ["Positivas", "Negativas"],
        "Cantidad": [positivas, negativas]
    })

    fig_distribucion = px.pie(
        distribucion,
        names="Tipo",
        values="Cantidad",
        hole=0.45
    )

    st.plotly_chart(
        fig_distribucion,
        use_container_width=True
    )

    st.markdown("**Juegos con mayor participación**")

    ranking = (
        filtradas
        .groupby("name", as_index=False)
        .agg(
            cantidad_resenas=("recommendationid", "count"),
            positivas=("voted_up", lambda x: (x == True).sum()),
            negativas=("voted_up", lambda x: (x == False).sum())
        )
    )

    total = ranking["positivas"] + ranking["negativas"]

    ranking["porcentaje_positivo"] = (
        ranking["positivas"]
        .div(total.where(total > 0))
        .mul(100)
    )

    ranking = ranking.sort_values(
        "cantidad_resenas",
        ascending=False
    )

    st.dataframe(
        ranking.head(10),
        use_container_width=True,
        hide_index=True
    )

    st.markdown("**Muestra de reseñas**")

    columnas = [
        c for c in [
            "account_created_year",
            "name",
            "voted_up",
            "review"
        ]
        if c in filtradas.columns
    ]

    st.dataframe(
        filtradas[columnas].head(100),
        use_container_width=True,
        hide_index=True
    )

# ============================================================
# 9. DIAGNOSTICO EMPRESARIAL
# ============================================================

st.divider()

st.header("Diagnóstico empresarial")

mayor_volumen = resumen.loc[
    resumen["cantidad_resenas"].idxmax()
]

mayor_aceptacion = resumen.loc[
    resumen["porcentaje_positivo"].idxmax()
]

menor_aceptacion = resumen.loc[
    resumen["porcentaje_positivo"].idxmin()
]

mayor_actividad = tabla.loc[
    tabla["resenas_por_usuario"].idxmax()
]

col1, col2 = st.columns(2)

with col1:

    st.markdown("**Mayor participación**")

    st.metric(
        f"Cohorte {int(mayor_volumen['account_created_year'])}",
        f"{int(mayor_volumen['cantidad_resenas']):,} reseñas"
    )

with col2:

    st.markdown("**Mayor aceptación**")

    st.metric(
        f"Cohorte {int(mayor_aceptacion['account_created_year'])}",
        f"{mayor_aceptacion['porcentaje_positivo']:.2f}%"
    )

st.write(
    f"La menor aceptación corresponde a la cohorte "
    f"{int(menor_aceptacion['account_created_year'])}, con "
    f"{menor_aceptacion['porcentaje_positivo']:.2f}%."
)

st.write(
    f"La mayor actividad promedio corresponde a la cohorte "
    f"{int(mayor_actividad['account_created_year'])}, con "
    f"{mayor_actividad['resenas_por_usuario']:.2f} reseñas por usuario."
)

# ============================================================
# 10. RECOMENDACIONES ESTRATEGICAS
# ============================================================

st.divider()

st.subheader("Recomendaciones estratégicas")

tab1, tab2, tab3 = st.tabs([
    "Desarrolladores",
    "Publishers",
    "Inversores"
])

with tab1:

    st.markdown(
        """
        - Investigar diferencias de aceptación entre cohortes.
        - Analizar problemas recurrentes por segmento.
        - Identificar géneros y atributos con mayor participación.
        - Evaluar mejoras de producto según las necesidades detectadas.
        """
    )

with tab2:

    st.markdown(
        """
        - Segmentar campañas según patrones de participación.
        - Identificar cohortes con mayor actividad observada.
        - Evaluar estrategias de comunidad y fidelización.
        - Contrastar el interés con datos de conversión y ventas.
        """
    )

with tab3:

    st.markdown(
        """
        - Utilizar la aceptación como señal complementaria de demanda.
        - Investigar la evolución de participación por generación.
        - Contrastar la información con indicadores financieros.
        - Evaluar riesgos de producto y oportunidades de mercado.
        """
    )

st.warning(
    """
    Las cohortes no representan directamente ventas, ingresos,
    retención ni beneficios. Para evaluar rentabilidad se necesitan
    datos comerciales, financieros y temporales adicionales.
    """
)

# ============================================================
# 11. RESUMEN EJECUTIVO
# ============================================================

st.divider()

st.header("Resumen ejecutivo")

col1, col2, col3 = st.columns(3)

col1.metric(
    "Mayor volumen",
    str(int(mayor_volumen["account_created_year"]))
)

col2.metric(
    "Mayor aceptación",
    str(int(mayor_aceptacion["account_created_year"]))
)

col3.metric(
    "Mayor actividad",
    str(int(mayor_actividad["account_created_year"]))
)

st.markdown(
    f"""
    El análisis principal comprende **{total_resenas:,} reseñas**
    de las cohortes 2020–2025.

    La aceptación positiva general es del
    **{porcentaje_positivo:.2f}%**.

    La cohorte con mayor volumen es
    **{int(mayor_volumen['account_created_year'])}**,
    mientras que la mayor aceptación corresponde a
    **{int(mayor_aceptacion['account_created_year'])}**.

    Estos resultados permiten orientar investigaciones
    comerciales y decisiones de producto.
    """
)

st.caption(
    "Steam Analytics | Cohortes principales 2020–2025."
)
'''

(PAGES_DIR / "02_Analisis_Cohortes.py").write_text(
    pagina_cohortes.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# PAGINA 3: EXPLORACION DE JUEGOS
# ============================================================

pagina_juegos = '''
import streamlit as st
import pandas as pd
import plotly.express as px

from utils import mostrar_navegacion
from utils.data_loader import cargar_datasets
from utils.visualizaciones import grafico_ranking_juegos

st.set_page_config(
    page_title="Exploración de juegos",
    page_icon="🎮",
    layout="wide"
)

mostrar_navegacion()

# ============================================================
# 1. CARGA DE DATOS
# ============================================================

datos = cargar_datasets()

ranking_base = datos["ranking_juegos"].copy()
resenas = datos["resenas_dashboard"].copy()

resenas["fecha_analisis"] = pd.to_datetime(
    resenas["timestamp_created_parsed"],
    errors="coerce",
    utc=True
)

st.title("🎮 Exploración de videojuegos")

st.markdown(
    """
    Evaluación individual y comparativa de videojuegos.

    Identificamos oportunidades de posicionamiento, aceptación,
    participación y posibles problemas de producto.
    """
)

if ranking_base.empty or resenas.empty:
    st.warning("No hay información suficiente para explorar videojuegos.")
    st.stop()

# ============================================================
# 2. FILTROS DE EXPLORACION
# ============================================================

st.divider()

st.subheader("Filtros")

juegos = sorted(
    ranking_base["name"].dropna().unique().tolist()
)

busqueda = st.text_input(
    "Buscar videojuego",
    placeholder="Ingresar nombre del videojuego..."
)

if busqueda.strip():
    juegos_filtrados = [
        nombre for nombre in juegos
        if busqueda.lower() in nombre.lower()
    ]
else:
    juegos_filtrados = juegos

if not juegos_filtrados:
    st.warning("No se encontraron videojuegos con ese nombre.")
    st.stop()

seleccion = st.selectbox(
    "Seleccionar videojuego",
    juegos_filtrados
)

# ============================================================
# 3. INDICADORES DEL JUEGO
# ============================================================

detalle = resenas[
    resenas["name"] == seleccion
].copy()

if detalle.empty:
    st.warning("No hay reseñas disponibles para este videojuego.")
    st.stop()

total_resenas = len(detalle)

valoraciones = detalle["voted_up"].dropna()

positivas = int((valoraciones == True).sum())
negativas = int((valoraciones == False).sum())

total_valoraciones = len(valoraciones)

porcentaje_positivo = (
    positivas / total_valoraciones * 100
    if total_valoraciones else 0
)

porcentaje_negativo = (
    negativas / total_valoraciones * 100
    if total_valoraciones else 0
)

usuarios = detalle["steamid"].nunique()

col1, col2, col3, col4 = st.columns(4)

col1.metric("Reseñas", f"{total_resenas:,}")
col2.metric("Usuarios", f"{usuarios:,}")
col3.metric("Positivas", f"{porcentaje_positivo:.2f}%")
col4.metric("Negativas", f"{porcentaje_negativo:.2f}%")

# ============================================================
# 4. DISTRIBUCION DE VALORACIONES
# ============================================================

st.divider()

st.subheader("Aceptación del videojuego")

col1, col2 = st.columns(2)

with col1:

    distribucion = pd.DataFrame({
        "Tipo": ["Positivas", "Negativas"],
        "Cantidad": [positivas, negativas]
    })

    fig = px.pie(
        distribucion,
        names="Tipo",
        values="Cantidad",
        hole=0.45,
        title="Distribución de recomendaciones"
    )

    st.plotly_chart(
        fig,
        use_container_width=True
    )

with col2:

    st.metric(
        "Total de valoraciones válidas",
        f"{total_valoraciones:,}"
    )

    st.metric(
        "Usuarios que reseñaron",
        f"{usuarios:,}"
    )

    st.metric(
        "Promedio de reseñas por usuario",
        f"{total_resenas / usuarios:.2f}" if usuarios else "0"
    )

# ============================================================
# 5. EVOLUCION TEMPORAL
# ============================================================

st.divider()

st.subheader("Evolución temporal de reseñas")

temporal = (
    detalle.dropna(subset=["fecha_analisis"])
    .assign(
        periodo=lambda x: x["fecha_analisis"].dt.strftime("%Y-%m")
    )
    .groupby("periodo", as_index=False)
    .size()
    .rename(columns={"size": "cantidad_resenas"})
)

if not temporal.empty:

    fig_temporal = px.line(
        temporal,
        x="periodo",
        y="cantidad_resenas",
        markers=True,
        title="Evolución mensual de reseñas",
        labels={
            "periodo": "Mes",
            "cantidad_resenas": "Reseñas"
        }
    )

    st.plotly_chart(
        fig_temporal,
        use_container_width=True
    )

    pico = temporal.loc[
        temporal["cantidad_resenas"].idxmax()
    ]

    st.caption(
        f"Mes con mayor actividad: {pico['periodo']} "
        f"({pico['cantidad_resenas']:,} reseñas)."
    )

# ============================================================
# 6. PARTICIPACION POR IDIOMA
# ============================================================

st.divider()

st.subheader("Distribución por idioma")

if "language_code" in detalle.columns:

    idiomas = (
        detalle["language_code"]
        .fillna("No especificado")
        .value_counts()
        .head(10)
        .rename_axis("Idioma")
        .reset_index(name="Reseñas")
    )

    st.dataframe(
        idiomas,
        use_container_width=True,
        hide_index=True
    )

# ============================================================
# 7. COMPARACION CON EL RESTO DE LOS JUEGOS
# ============================================================

st.divider()

st.subheader("Posicionamiento comparativo")

ranking = (
    resenas
    .groupby("name", as_index=False)
    .agg(
        cantidad_resenas=("recommendationid", "count"),
        positivas=("voted_up", lambda x: (x == True).sum()),
        negativas=("voted_up", lambda x: (x == False).sum()),
        total_valoraciones=("voted_up", "count")
    )
)

ranking["porcentaje_positivo"] = (
    ranking["positivas"]
    .div(ranking["total_valoraciones"].where(
        ranking["total_valoraciones"] > 0
    ))
    .mul(100)
)

ranking = ranking.sort_values(
    "cantidad_resenas",
    ascending=False
).reset_index(drop=True)

posicion = ranking[
    ranking["name"] == seleccion
].index

if len(posicion):

    posicion = int(posicion[0]) + 1

    st.metric(
        "Posición por volumen de reseñas",
        f"{posicion} / {len(ranking)}"
    )

    promedio_mercado = ranking[
        "porcentaje_positivo"
    ].mean()

    diferencia = porcentaje_positivo - promedio_mercado

    st.metric(
        "Diferencia frente al promedio de la muestra",
        f"{diferencia:+.2f} puntos porcentuales"
    )

# ============================================================
# 8. DIAGNOSTICO DEL VIDEOJUEGO
# ============================================================

st.divider()

st.header("Diagnóstico empresarial")

if total_valoraciones < 30:

    st.info(
        "Muestra reducida: se recomienda ampliar la observación "
        "antes de formular conclusiones comerciales."
    )

elif porcentaje_negativo >= 30:

    st.error(
        f"""
        Señal de riesgo: el {porcentaje_negativo:.2f}% de las
        valoraciones son negativas.

        Recomendación: investigar problemas de rendimiento,
        contenido, jugabilidad y experiencia del usuario.
        """
    )

elif porcentaje_positivo >= 80:

    st.success(
        f"""
        Señal favorable: el {porcentaje_positivo:.2f}% de las
        valoraciones son positivas.

        Recomendación: identificar los atributos mejor valorados
        y estudiar oportunidades de expansión y posicionamiento.
        """
    )

else:

    st.warning(
        "Aceptación intermedia: analizar las reseñas para "
        "detectar factores de satisfacción e insatisfacción."
    )

# ============================================================
# 9. OPORTUNIDADES DE MEJORA
# ============================================================

st.divider()

st.subheader("Recomendaciones por público")

tab1, tab2, tab3 = st.tabs([
    "Desarrolladores",
    "Publishers",
    "Inversores"
])

with tab1:

    st.markdown(
        """
        - Priorizar problemas recurrentes de las reseñas.
        - Analizar cambios en la percepción a lo largo del tiempo.
        - Investigar oportunidades de mejora de producto.
        - Medir la respuesta de la comunidad a las actualizaciones.
        """
    )

with tab2:

    st.markdown(
        """
        - Evaluar el posicionamiento frente a otros juegos.
        - Identificar mercados e idiomas con participación.
        - Contrastar el interés con conversión y ventas.
        - Investigar oportunidades de comunicación comercial.
        """
    )

with tab3:

    st.markdown(
        """
        - Considerar aceptación y participación como señales complementarias.
        - Evaluar riesgos reputacionales.
        - Contrastar resultados con ingresos y costos.
        - Analizar la evolución de la demanda con datos externos.
        """
    )

st.warning(
    """
    El volumen de reseñas no representa ventas ni ingresos.
    La aceptación tampoco demuestra rentabilidad.
    Se requieren datos económicos adicionales para evaluar
    el desempeño financiero del videojuego.
    """
)

# ============================================================
# 10. RESEÑAS DEL VIDEOJUEGO
# ============================================================

st.divider()

st.subheader("Explorar reseñas")

tipo = st.selectbox(
    "Filtrar reseñas",
    ["Todas", "Positivas", "Negativas"]
)

reseñas = detalle.copy()

if tipo == "Positivas":
    reseñas = reseñas[
        reseñas["voted_up"] == True
    ]

elif tipo == "Negativas":
    reseñas = reseñas[
        reseñas["voted_up"] == False
    ]

columnas = [
    c for c in [
        "timestamp_created_parsed",
        "language_code",
        "voted_up",
        "votes_up",
        "review"
    ]
    if c in reseñas.columns
]

st.dataframe(
    reseñas[columnas].head(100),
    use_container_width=True,
    hide_index=True
)
'''

(PAGES_DIR / "03_Exploracion_Juegos.py").write_text(
    pagina_juegos.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# PAGINA 4: ANALISIS DE SENTIMIENTO
# ============================================================

pagina_sentimiento = '''
import streamlit as st
import pandas as pd

from utils import mostrar_navegacion
from utils.data_loader import cargar_modelo, cargar_datasets
from utils.predicciones import predecir_sentimiento

st.set_page_config(
    page_title="Análisis de sentimiento",
    page_icon="💬",
    layout="wide"
)

mostrar_navegacion()

st.title("💬 Análisis de sentimiento")

st.markdown(
    """
    Análisis automatizado de opiniones de usuarios mediante
    TF-IDF y Linear SVM.

    El objetivo es identificar señales de satisfacción,
    problemas recurrentes y oportunidades de mejora de producto.
    """
)

# ============================================================
# 1. CARGA DEL MODELO
# ============================================================

try:
    modelo = cargar_modelo()
except Exception as error:
    st.error(
        f"No fue posible cargar el modelo: {error}"
    )
    st.stop()

# ============================================================
# 2. CLASIFICACION INDIVIDUAL
# ============================================================

st.divider()

st.subheader("Analizar una reseña")

texto = st.text_area(
    "Ingresar una reseña en inglés",
    height=180,
    placeholder="Write your game review here..."
)

if st.button(
    "Analizar reseña",
    type="primary"
):

    if not texto.strip():

        st.warning("Ingresá una reseña antes de analizar.")

    else:

        try:

            resultado = predecir_sentimiento(
                modelo,
                texto
            )

            sentimiento = resultado["sentimiento"]

            if sentimiento == "Positiva":

                st.success(
                    "Sentimiento detectado: POSITIVO"
                )

            else:

                st.error(
                    "Sentimiento detectado: NEGATIVO"
                )

            margen = resultado.get(
                "margen_decision"
            )

            if margen is not None:

                st.metric(
                    "Margen de decisión",
                    f"{margen:.4f}"
                )

                st.caption(
                    "El margen de decisión no es una probabilidad."
                )

            st.subheader("Interpretación empresarial")

            if sentimiento == "Positiva":

                st.markdown(
                    """
                    **Señal favorable**

                    La reseña contiene una opinión clasificada
                    como positiva por el modelo.

                    Recomendación: identificar atributos valorados
                    y analizar si aparecen de forma recurrente
                    en otras opiniones.
                    """
                )

            else:

                st.markdown(
                    """
                    **Señal de atención**

                    La reseña contiene una opinión clasificada
                    como negativa por el modelo.

                    Recomendación: investigar los problemas
                    descritos y determinar si requieren mejoras
                    de producto o comunicación.
                    """
                )

        except (TypeError, ValueError) as error:

            st.warning(str(error))

        except Exception as error:

            st.error(
                f"Error durante la predicción: {error}"
            )

# ============================================================
# 3. ANALISIS DE RESEÑAS DE LA BASE
# ============================================================

st.divider()

st.subheader("Exploración de reseñas disponibles")

st.markdown(
    """
    Esta sección permite consultar las reseñas reales de la muestra.
    La clasificación individual del modelo y la recomendación original
    de Steam son indicadores diferentes.
    """
)

try:

    datos = cargar_datasets()

    resenas = datos["resenas_dashboard"].copy()

    if not resenas.empty:

        if "voted_up" in resenas.columns:

            positivas = int(
                resenas["voted_up"].eq(True).sum()
            )

            negativas = int(
                resenas["voted_up"].eq(False).sum()
            )

            total = positivas + negativas

            porcentaje_positivo = (
                positivas / total * 100
                if total else 0
            )

            col1, col2, col3 = st.columns(3)

            col1.metric(
                "Reseñas",
                f"{len(resenas):,}"
            )

            col2.metric(
                "Recomendaciones positivas",
                f"{porcentaje_positivo:.2f}%"
            )

            col3.metric(
                "Recomendaciones negativas",
                f"{negativas:,}"
            )

        # Filtros de consulta
        juegos = sorted(
            resenas["name"].dropna().unique().tolist()
        )

        juego = st.selectbox(
            "Filtrar por videojuego",
            ["Todos"] + juegos,
            key="sentimiento_juego"
        )

        tipo = st.selectbox(
            "Filtrar por recomendación original",
            ["Todas", "Positivas", "Negativas"],
            key="sentimiento_tipo"
        )

        filtradas = resenas.copy()

        if juego != "Todos":

            filtradas = filtradas[
                filtradas["name"] == juego
            ]

        if tipo == "Positivas":

            filtradas = filtradas[
                filtradas["voted_up"] == True
            ]

        elif tipo == "Negativas":

            filtradas = filtradas[
                filtradas["voted_up"] == False
            ]

        st.write(
            f"Reseñas encontradas: {len(filtradas):,}"
        )

        columnas = [
            c for c in [
                "name",
                "voted_up",
                "language_code",
                "review"
            ]
            if c in filtradas.columns
        ]

        st.dataframe(
            filtradas[columnas].head(100),
            use_container_width=True,
            hide_index=True
        )

    else:

        st.info("No hay reseñas disponibles.")

except Exception as error:

    st.warning(
        f"No fue posible cargar la exploración de reseñas: {error}"
    )

# ============================================================
# 4. RECOMENDACIONES DE USO DEL MODELO
# ============================================================

st.divider()

st.header("Aplicación empresarial del análisis de sentimiento")

tab1, tab2, tab3 = st.tabs([
    "Desarrolladores",
    "Publishers",
    "Inversores"
])

with tab1:

    st.markdown(
        """
        - Detectar problemas de experiencia.
        - Priorizar mejoras en función de las opiniones.
        - Monitorear cambios en la percepción después de actualizaciones.
        - Identificar atributos valorados por la comunidad.
        """
    )

with tab2:

    st.markdown(
        """
        - Monitorear riesgos reputacionales.
        - Identificar señales de aceptación o rechazo.
        - Evaluar la respuesta de la comunidad a campañas y lanzamientos.
        - Complementar estudios de mercado.
        """
    )

with tab3:

    st.markdown(
        """
        - Utilizar el sentimiento como indicador complementario.
        - Investigar riesgos de producto y reputación.
        - Contrastar las señales con métricas comerciales.
        - Evaluar tendencias de aceptación antes de tomar decisiones.
        """
    )

st.warning(
    """
    El modelo clasifica texto según patrones aprendidos durante
    el entrenamiento. No garantiza una interpretación perfecta,
    no identifica causalidad y no estima ventas ni rentabilidad.
    """
)

st.caption(
    "Modelo de clasificación: TF-IDF + Linear SVM."
)
'''

(PAGES_DIR / "04_Analisis_Sentimiento.py").write_text(
    pagina_sentimiento.strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# 3. VERIFICACION DE SINTAXIS Y CONTENIDO
# ============================================================

print("\n" + "=" * 70)
print("VERIFICACION DE PAGINAS STREAMLIT")
print("=" * 70)

archivos_verificar = [
    UTILS_DIR / "__init__.py",
    *sorted(PAGES_DIR.glob("*.py"))
]

errores = []

for archivo in archivos_verificar:

    contenido = archivo.read_text(
        encoding="utf-8"
    )

    try:

        ast.parse(contenido)

        estado = "SINTAXIS OK"

    except SyntaxError as error:

        estado = (
            f"ERROR LINEA {error.lineno}: {error.msg}"
        )

        errores.append(
            (archivo.name, error)
        )

    print(
        f"{archivo.name} | "
        f"{archivo.stat().st_size:,} bytes | "
        f"{estado}"
    )

print("\n" + "=" * 70)

if errores:

    print("SE DETECTARON ERRORES DE SINTAXIS.")

    for nombre, error in errores:

        print(
            f"{nombre}: línea {error.lineno} - {error.msg}"
        )

else:

    print("TODOS LOS ARCHIVOS VERIFICADOS CORRECTAMENTE.")

# ============================================================
# 4. VERIFICACION DE INTEGRIDAD DE LAS CUATRO PAGINAS
# ============================================================

print("\n" + "=" * 70)
print("VERIFICACION DE INTEGRIDAD DEL CONTENIDO")
print("=" * 70)

marcadores = {
    "01_Analisis_General.py": [
        "Diagnóstico empresarial",
        "Ranking de videojuegos",
        "Recomendaciones empresariales"
    ],
    "02_Analisis_Cohortes.py": [
        "ANIOS_COHORTE",
        "Comparación detallada",
        "Resumen ejecutivo"
    ],
    "03_Exploracion_Juegos.py": [
        "Posicionamiento comparativo",
        "Evolución temporal de reseñas",
        "Diagnóstico empresarial"
    ],
    "04_Analisis_Sentimiento.py": [
        "Analizar una reseña",
        "Linear SVM",
        "Aplicación empresarial"
    ]
}

for nombre, textos in marcadores.items():

    archivo = PAGES_DIR / nombre

    contenido = archivo.read_text(
        encoding="utf-8"
    )

    faltantes = [
        texto for texto in textos
        if texto not in contenido
    ]

    if faltantes:

        print(
            f"{nombre}: FALTAN MARCADORES {faltantes}"
        )

        errores.append(
            (nombre, "Contenido incompleto")
        )

    else:

        print(
            f"{nombre}: CONTENIDO COMPLETO"
        )

print("\n" + "=" * 70)

if errores:

    print("VERIFICAR LOS ERRORES ANTES DE CONTINUAR.")

else:

    print("PASO 7 FINALIZADO CORRECTAMENTE.")
    print("Las cuatro páginas fueron generadas y verificadas.")
````

## 10. Módulo de carga de datos

Se desarrolla un módulo reutilizable para:

- Cargar los seis datasets Parquet.
- Cargar el modelo predictivo.
- Centralizar las rutas de los recursos.
- Utilizar caché de Streamlit para optimizar tiempos de carga.

Este módulo será utilizado por todas las páginas de la aplicación.

````python
# ============================================================
# PASO 10: MODULO DATA LOADER
# ============================================================

DATA_LOADER_CODE = '''
from pathlib import Path

import joblib
import pandas as pd
import streamlit as st


# ------------------------------------------------------------
# RUTAS
# ------------------------------------------------------------

BASE_DIR = Path(__file__).resolve().parent.parent
DATA_DIR = BASE_DIR / "Data"
MODELS_DIR = BASE_DIR / "models"

MODEL_PATH = MODELS_DIR / "steam_linear_svm_tfidf_final.joblib"


# ------------------------------------------------------------
# CARGA DE DATASETS
# ------------------------------------------------------------

@st.cache_data
def cargar_datasets():

    archivos = {
        "resenas_dashboard": "resenas_dashboard.parquet",
        "resenas_cohortes": "resenas_cohortes.parquet",
        "kpis_generales": "kpis_generales.parquet",
        "resumen_cohortes": "resumen_cohortes.parquet",
        "actividad_mensual": "actividad_mensual.parquet",
        "ranking_juegos": "ranking_juegos.parquet"
    }

    datos = {}

    for nombre, archivo in archivos.items():

        ruta = DATA_DIR / archivo

        if not ruta.exists():
            raise FileNotFoundError(
                f"No se encontro el archivo: {ruta}"
            )

        datos[nombre] = pd.read_parquet(ruta)

    return datos


# ------------------------------------------------------------
# CARGA DEL MODELO
# ------------------------------------------------------------

@st.cache_resource
def cargar_modelo():

    if not MODEL_PATH.exists():
        raise FileNotFoundError(
            f"No se encontro el modelo: {MODEL_PATH}"
        )

    return joblib.load(MODEL_PATH)
'''

DATA_LOADER_PATH = UTILS_DIR / "data_loader.py"

DATA_LOADER_PATH.write_text(
    DATA_LOADER_CODE.strip() + "\n",
    encoding="utf-8"
)

print("MODULO DATA LOADER CREADO")
print("=" * 60)
print(DATA_LOADER_PATH)
print(f"Tamano: {DATA_LOADER_PATH.stat().st_size} bytes")
````

## 14. Módulo de predicciones

Se desarrolla una función para clasificar nuevas reseñas mediante el Pipeline TF-IDF + Linear SVM previamente entrenado.

La aplicación no realizará entrenamiento ni ajuste del modelo.

El margen de decisión se mostrará como información complementaria y no será interpretado como una probabilidad.

````python
# ============================================================
# PASO 14: MODULO DE PREDICCIONES
# ============================================================

PREDICCIONES_CODE = '''
def predecir_sentimiento(modelo, texto):

    if not isinstance(texto, str):
        raise TypeError("La reseña debe ser un texto.")

    texto = texto.strip()

    if not texto:
        raise ValueError("La reseña no puede estar vacía.")

    prediccion = modelo.predict([texto])[0]

    resultado = {
        1: "Positiva",
        0: "Negativa",
        True: "Positiva",
        False: "Negativa"
    }

    sentimiento = resultado.get(prediccion, str(prediccion))

    margen = None

    if hasattr(modelo, "decision_function"):
        margen = float(modelo.decision_function([texto])[0])

    return {
        "prediccion": prediccion,
        "sentimiento": sentimiento,
        "margen_decision": margen
    }
'''

PREDICCIONES_PATH = UTILS_DIR / "predicciones.py"

PREDICCIONES_PATH.write_text(
    PREDICCIONES_CODE.strip() + "\n",
    encoding="utf-8"
)

print("MODULO DE PREDICCIONES CREADO")
print(PREDICCIONES_PATH)
print(f"Tamaño: {PREDICCIONES_PATH.stat().st_size} bytes")
````
